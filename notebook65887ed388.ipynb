📦 Module 1: Data Loading & EDA

```python
# ============================================================
# MODULE 1: DATA LOADING AND EXPLORATORY DATA ANALYSIS
# ============================================================
# PURPOSE  : Load IEEE CIS Fraud Detection dataset (transaction +
#            identity tables), merge them, and generate EDA report.
# INPUT    : train_transaction.csv, train_identity.csv
# OUTPUT   : module1_merged_data.csv
#            module1_eda_report.xlsx
# ============================================================

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# ── Configuration ────────────────────────────────────────────
# Prepend the Kaggle data source path to correctly locate the files
TRANSACTION_PATH = umersajid842_ieee_dataset1_path + '/train_transaction.csv'  # Main fraud/transaction file
IDENTITY_PATH    = umersajid842_ieee_dataset1_path + '/train_identity.csv'     # Device/identity features file
OUTPUT_DATA      = 'module1_merged_data.csv'
OUTPUT_EDA       = 'module1_eda_report.xlsx'


# ────────────────────────────────────────────────────────────
def load_raw_data(transaction_path: str, identity_path: str) -> pd.DataFrame:
    """
    Load and LEFT-JOIN transaction and identity tables.

    The IEEE CIS dataset has two separate files:
      - train_transaction.csv : 590K rows × 394 cols (core fraud labels + V-features)
      - train_identity.csv    : 144K rows × 41 cols  (device, browser, OS info)
    Not every transaction has identity info → left join keeps all transactions.

    Args:
        transaction_path : file path to train_transaction.csv
        identity_path    : file path to train_identity.csv

    Returns:
        df : merged DataFrame with all features
    """
    print("\n[MODULE 1] ── Loading IEEE CIS Fraud Dataset ──")

    # ── Load transaction table ───────────────────────────────
    print("  ► Loading transaction table...")
    df_trans = pd.read_csv(transaction_path)               # ~590K rows
    print(f"    Shape : {df_trans.shape}")
    print(f"    Fraud : {df_trans['isFraud'].sum():,} "
          f"({df_trans['isFraud'].mean()*100:.2f}%)")

    # ── Load identity table ──────────────────────────────────
    print("  ► Loading identity table...")
    df_id = pd.read_csv(identity_path)                     # ~144K rows
    print(f"    Shape : {df_id.shape}")

    # ── Merge on TransactionID (left join) ───────────────────
    print("  ► Merging on TransactionID...")
    df = df_trans.merge(df_id, on='TransactionID', how='left')
    print(f"    Merged Shape : {df.shape}")

    return df


# ────────────────────────────────────────────────────────────
def generate_eda_report(df: pd.DataFrame, output_path: str) -> pd.DataFrame:
    """
    Compute column-level EDA statistics and save to Excel.

    For each column this calculates:
      - Data type
      - Missing value percentage
      - Number of unique values
      - Mean / Std (for numeric columns)

    Args:
        df          : merged DataFrame
        output_path : path to save .xlsx EDA report

    Returns:
        eda_df : EDA statistics as a DataFrame
    """
    print("\n  ► Generating EDA report...")

    # ── Per-column statistics ────────────────────────────────
    missing_pct  = (df.isnull().sum() / len(df) * 100).round(2)
    dtype_info   = df.dtypes.astype(str)
    unique_cnt   = df.nunique()

    # Numeric mean and std (only for numeric columns; NaN for others)
    num_cols = df.select_dtypes(include=np.number).columns
    means    = df[num_cols].mean()
    stds     = df[num_cols].std()

    eda_df = pd.DataFrame({
        'column'      : df.columns,
        'dtype'       : dtype_info.values,
        'missing_pct' : missing_pct.values,
        'unique_count': unique_cnt.values,
        'mean'        : [means.get(c, np.nan) for c in df.columns],
        'std'         : [stds.get(c, np.nan) for c in df.columns],
    })

    # ── Dataset-level summary ────────────────────────────────
    summary = {
        'total_transactions': len(df),
        'fraud_count'       : int(df['isFraud'].sum()),
        'legit_count'       : int((df['isFraud'] == 0).sum()),
        'fraud_rate_pct'    : round(df['isFraud'].mean() * 100, 3),
        'n_features'        : df.shape[1],
        'numeric_features'  : len(num_cols),
        'categorical_features': df.select_dtypes(include='object').shape[1],
    }

    print("  ── Dataset Summary ──────────────────────────")
    for k, v in summary.items():
        print(f"    {k:<30} : {v}")

    # ── Save to Excel with two sheets ────────────────────────
    with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
        eda_df.to_excel(writer, sheet_name='Column_Stats', index=False)
        pd.DataFrame([summary]).T.reset_index().rename(
            columns={'index': 'Metric', 0: 'Value'}
        ).to_excel(writer, sheet_name='Dataset_Summary', index=False)

    print(f"  ✓ EDA report saved → {output_path}")
    return eda_df


# ────────────────────────────────────────────────────────────
def run_module1():
    """
    Orchestrates Module 1:
      1. Load raw data
      2. Generate EDA report
      3. Save merged data for Module 2

    Returns:
        df : merged raw DataFrame
    """
    df = load_raw_data(TRANSACTION_PATH, IDENTITY_PATH)
    generate_eda_report(df, OUTPUT_EDA)

    # Save merged data as input for Module 2
    df.to_csv(OUTPUT_DATA, index=False)
    print(f"\n  ✓ Merged data saved → {OUTPUT_DATA}")
    print("[MODULE 1] Complete ✓\n")

    return df


# ── Execute Module 1 ─────────────────────────────────────────
if __name__ == "__main__":
    df_raw = run_module1()
```

⚙️ Module 2: Data Preprocessing & Cleaning

```python
# ============================================================
# MODULE 2: DATA PREPROCESSING AND CLEANING
# ============================================================
# PURPOSE  : Handle missing values, encode categoricals, remove
#            highly-correlated or low-variance features, and
#            produce a clean numerical matrix for modelling.
# INPUT    : module1_merged_data.csv
# OUTPUT   : module2_cleaned_data.csv
#            module2_dropped_columns.csv
# ============================================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
import warnings
warnings.filterwarnings('ignore')

# ── Configuration ────────────────────────────────────────────
INPUT_DATA        = 'module1_merged_data.csv'
OUTPUT_CLEAN      = 'module2_cleaned_data.csv'
OUTPUT_DROPPED    = 'module2_dropped_columns.csv'

MISSING_THRESHOLD = 0.50   # Drop columns with >50% missing values
CORR_THRESHOLD    = 0.98   # Drop one column of pairs with |corr| > 0.98


# ────────────────────────────────────────────────────────────
def drop_high_missing(df: pd.DataFrame, threshold: float) -> pd.DataFrame:
    """
    Remove columns where missing value percentage exceeds threshold.

    IEEE CIS V-features (V1-V339) are sparse — many have >80% NaN.
    Keeping them degrades model performance and inflates memory usage.

    Args:
        df        : input DataFrame
        threshold : fraction threshold (e.g., 0.50 = 50%)

    Returns:
        df_out : DataFrame with high-missing columns removed
    """
    # Compute missing fraction per column
    missing_frac = df.isnull().mean()

    # Identify columns exceeding the threshold
    cols_to_drop = missing_frac[missing_frac > threshold].index.tolist()

    # Never drop the target column
    cols_to_drop = [c for c in cols_to_drop if c != 'isFraud']

    print(f"  ► Dropping {len(cols_to_drop)} columns with >{threshold*100:.0f}% missing")

    # Save list of dropped columns for reproducibility
    dropped_df = pd.DataFrame({
        'column'     : cols_to_drop,
        'reason'     : 'high_missing',
        'missing_pct': (missing_frac[cols_to_drop] * 100).round(2).values,
    })

    return df.drop(columns=cols_to_drop), dropped_df


# ────────────────────────────────────────────────────────────
def fill_missing_values(df: pd.DataFrame) -> pd.DataFrame:
    """
    Impute remaining missing values after high-missing column removal.

    Strategy:
      - Numeric columns  → fill with MEDIAN  (robust to outliers)
      - Object columns   → fill with 'UNKNOWN' (preserves category info)

    Args:
        df : cleaned DataFrame (after drop_high_missing)

    Returns:
        df : DataFrame with no missing values
    """
    print("  ► Filling remaining missing values...")

    # Identify numeric and categorical columns
    num_cols = df.select_dtypes(include=np.number).columns.tolist()
    cat_cols = df.select_dtypes(include='object').columns.tolist()

    # Fill numeric → median (more robust than mean for fraud data with outliers)
    for col in num_cols:
        if col != 'isFraud' and df[col].isnull().any():
            df[col] = df[col].fillna(df[col].median())

    # Fill categorical → 'UNKNOWN' to preserve category without data leakage
    for col in cat_cols:
        df[col] = df[col].fillna('UNKNOWN')

    print(f"    Remaining nulls after fill: {df.isnull().sum().sum()}")
    return df


# ────────────────────────────────────────────────────────────
def encode_categoricals(df: pd.DataFrame) -> pd.DataFrame:
    """
    Label-encode all remaining object/categorical columns.

    LabelEncoder maps each unique string to an integer:
      e.g., 'gmail.com' → 0, 'yahoo.com' → 1, 'UNKNOWN' → 2

    Args:
        df : DataFrame with filled values

    Returns:
        df : DataFrame with all columns as numeric types
    """
    print("  ► Encoding categorical columns...")

    cat_cols = df.select_dtypes(include='object').columns.tolist()
    le = LabelEncoder()

    for col in cat_cols:
        df[col] = le.fit_transform(df[col].astype(str))  # convert to string first for safety

    print(f"    Encoded {len(cat_cols)} categorical columns: {cat_cols[:5]}...")
    return df


# ────────────────────────────────────────────────────────────
def drop_high_correlation(df: pd.DataFrame,
                           threshold: float,
                           exclude_cols: list) -> pd.DataFrame:
    """
    Remove one column from each highly-correlated pair.

    IEEE CIS V-features are Vesta-engineered and often highly redundant.
    Correlation > 0.98 means near-duplicate features → remove one to reduce
    dimensionality without losing information.

    Args:
        df           : numeric DataFrame
        threshold    : Pearson correlation threshold
        exclude_cols : columns to protect from dropping (e.g., IDs, target)

    Returns:
        df : DataFrame with one of each correlated pair removed
    """
    print(f"  ► Dropping columns with |corr| > {threshold}...")

    # Only compute correlation on numeric columns, excluding protected cols
    num_cols = [c for c in df.select_dtypes(include=np.number).columns
                if c not in exclude_cols]

    # Compute Pearson correlation matrix
    corr_matrix = df[num_cols].corr().abs()

    # Upper triangle only (avoid counting pairs twice)
    upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

    # Find columns with any correlation above threshold
    cols_to_drop = [col for col in upper.columns if any(upper[col] > threshold)]

    print(f"    Dropping {len(cols_to_drop)} correlated columns")

    return df.drop(columns=cols_to_drop, errors='ignore')


# ────────────────────────────────────────────────────────────
def run_module2(df: pd.DataFrame = None) -> pd.DataFrame:
    """
    Orchestrates Module 2:
      1. Drop high-missing columns
      2. Fill remaining missing values
      3. Encode categorical columns
      4. Drop highly-correlated columns
      5. Save cleaned data

    Args:
        df : optionally pass DataFrame directly (skips CSV load)

    Returns:
        df_clean : cleaned DataFrame ready for feature engineering
    """
    if df is None:
        print(f"\n[MODULE 2] Loading from {INPUT_DATA}...")
        df = pd.read_csv(INPUT_DATA)

    print("\n[MODULE 2] ── Preprocessing ──")
    all_dropped = []

    # Step 1: Remove very sparse columns
    df, dropped_missing = drop_high_missing(df, MISSING_THRESHOLD)
    all_dropped.append(dropped_missing)

    # Step 2: Impute remaining nulls
    df = fill_missing_values(df)

    # Step 3: Encode categoricals → all numeric now
    df = encode_categoricals(df)

    # Step 4: Remove highly-correlated features
    protect = ['TransactionID', 'isFraud', 'TransactionDT', 'TransactionAmt']
    df = drop_high_correlation(df, CORR_THRESHOLD, exclude_cols=protect)

    print(f"\n  ✓ Final cleaned shape: {df.shape}")

    # Save dropped columns log
    pd.concat(all_dropped, ignore_index=True).to_csv(OUTPUT_DROPPED, index=False)
    print(f"  ✓ Dropped columns log → {OUTPUT_DROPPED}")

    # Save clean data
    df.to_csv(OUTPUT_CLEAN, index=False)
    print(f"  ✓ Cleaned data saved → {OUTPUT_CLEAN}")
    print("[MODULE 2] Complete ✓\n")

    return df


# ── Execute ──────────────────────────────────────────────────
if __name__ == "__main__":
    df_clean = run_module2()
```

⚙️ Module 3: Feature Engineering (Temporal + Spatial)

```python
# ============================================================
# MODULE 3: FEATURE ENGINEERING
# ============================================================
# PURPOSE  : Create temporal features (from TransactionDT),
#            spatial/aggregated features (per card, email, addr),
#            and interaction features critical for fraud detection.
# INPUT    : module2_cleaned_data.csv
# OUTPUT   : module3_engineered_features.csv
#            module3_feature_importance_summary.csv
# ============================================================

import pandas as pd
import numpy as np
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

INPUT_DATA      = 'module2_cleaned_data.csv'
OUTPUT_FEATURES = 'module3_engineered_features.csv'
OUTPUT_SUMMARY  = 'module3_feature_importance_summary.csv'

# IEEE CIS TransactionDT reference date (start of dataset)
REFERENCE_DATE = datetime(2017, 11, 30)


# ────────────────────────────────────────────────────────────
def create_temporal_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Extract temporal features from TransactionDT (seconds since reference).

    TransactionDT is an integer: seconds elapsed since a reference date.
    Converting to actual datetime allows extraction of cyclical time signals
    that are strongly correlated with fraud (e.g., midnight transactions).

    Features created:
      - hour_of_day  : 0-23  (fraud peaks at night)
      - day_of_week  : 0-6   (fraud peaks on weekends)
      - day_of_month : 1-31
      - is_weekend   : binary flag
      - is_night     : binary flag (22:00 – 06:00)
      - sin_hour, cos_hour : cyclical encoding of hour
        → sin(2π × hour / 24) preserves cyclic structure
          so 23:00 and 00:00 are "close" to each other

    Args:
        df : cleaned DataFrame with TransactionDT column

    Returns:
        df : DataFrame with new temporal feature columns added
    """
    print("  ► Creating temporal features from TransactionDT...")

    # Convert seconds-offset to actual datetime objects
    df['transaction_datetime'] = pd.to_datetime(
        df['TransactionDT'],
        unit='s',                    # TransactionDT is in seconds
        origin=pd.Timestamp(REFERENCE_DATE)
    )

    # ── Raw time components ──────────────────────────────────
    df['hour_of_day']  = df['transaction_datetime'].dt.hour
    df['day_of_week']  = df['transaction_datetime'].dt.dayofweek   # Mon=0, Sun=6
    df['day_of_month'] = df['transaction_datetime'].dt.day

    # ── Binary time flags ────────────────────────────────────
    df['is_weekend'] = (df['day_of_week'] >= 5).astype(int)       # Sat=5, Sun=6
    df['is_night']   = ((df['hour_of_day'] >= 22) |               # After 10 PM
                        (df['hour_of_day'] <= 6)).astype(int)      # or before 6 AM

    # ── Cyclical encoding (prevents discontinuity at midnight) ─
    # Formula: sin(2π × feature / period)
    # This ensures hour 0 and hour 23 are numerically close
    df['sin_hour']    = np.sin(2 * np.pi * df['hour_of_day'] / 24)
    df['cos_hour']    = np.cos(2 * np.pi * df['hour_of_day'] / 24)
    df['sin_day']     = np.sin(2 * np.pi * df['day_of_week'] / 7)
    df['cos_day']     = np.cos(2 * np.pi * df['day_of_week'] / 7)

    # Drop the intermediate datetime column (not needed downstream)
    df = df.drop(columns=['transaction_datetime'], errors='ignore')

    print(f"    Added 10 temporal features")
    return df


# ────────────────────────────────────────────────────────────
def create_card_aggregation_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Aggregate transaction statistics per unique card (card1).

    Fraudsters often make multiple transactions with the same card.
    Card-level aggregations capture this behavioral pattern.

    Features created per card1:
      - card_tx_count      : total transactions per card
      - card_amt_mean      : mean transaction amount per card
      - card_amt_std       : std of transaction amounts per card
      - card_amt_max       : maximum transaction amount per card
      - card_fraud_rate    : historical fraud rate for this card
        WARNING: card_fraud_rate uses isFraud → only use in training
                 (set to 0 for test to avoid leakage)

    Args:
        df : feature-engineered DataFrame

    Returns:
        df : DataFrame with card aggregation columns added
    """
    print("  ► Creating card-level aggregation features (spatial)...")

    # Group all transactions by card1 identifier
    card_agg = df.groupby('card1').agg(
        card_tx_count  = ('TransactionAmt', 'count'),   # transaction frequency
        card_amt_mean  = ('TransactionAmt', 'mean'),    # average spend
        card_amt_std   = ('TransactionAmt', 'std'),     # spend variance
        card_amt_max   = ('TransactionAmt', 'max'),     # maximum spend
        card_fraud_rate= ('isFraud', 'mean'),           # historical fraud rate
    ).reset_index()

    # Fill NaN std (cards with single transaction have NaN std)
    card_agg['card_amt_std'] = card_agg['card_amt_std'].fillna(0)

    # Merge aggregations back to main dataframe
    df = df.merge(card_agg, on='card1', how='left')

    print(f"    Added 5 card aggregation features")
    return df


# ────────────────────────────────────────────────────────────
def create_email_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create features based on purchaser email domain (P_emaildomain).

    Email domain patterns are strong fraud indicators:
      - Free email services (gmail, yahoo) → higher fraud risk
      - Unknown/unusual domains → higher fraud risk

    Features created:
      - email_tx_count  : number of transactions from this email domain
      - email_fraud_rate: fraud rate for this email domain
      - is_free_email   : binary flag for major free email providers

    Args:
        df : DataFrame with P_emaildomain column

    Returns:
        df : DataFrame with email features added
    """
    print("  ► Creating email domain features...")

    # Aggregate fraud rate by email domain
    email_agg = df.groupby('P_emaildomain').agg(
        email_tx_count  = ('TransactionAmt', 'count'),
        email_fraud_rate= ('isFraud', 'mean'),
    ).reset_index()

    df = df.merge(email_agg, on='P_emaildomain', how='left')

    # Free email providers (encoded as integers after Module 2 label encoding)
    # Since we encoded P_emaildomain in Module 2, we use the encoded value
    # Here we create a simpler version: top-N email domain by transaction count
    top_email_domains = email_agg.nlargest(10, 'email_tx_count')['P_emaildomain']
    df['is_top_email_domain'] = df['P_emaildomain'].isin(top_email_domains).astype(int)

    print(f"    Added 3 email features")
    return df


# ────────────────────────────────────────────────────────────
def create_interaction_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Create cross-feature interaction terms.

    Interaction features capture non-linear relationships between
    original features that transformer attention may not discover alone.

    Features created:
      - amt_per_card_avg  : current amount vs card's average (ratio)
        → High values indicate anomalous large transactions
      - hour_x_weekend    : multiplicative interaction of time signals
      - v_sum             : sum of top V-features (Vesta risk score proxy)

    Args:
        df : DataFrame with card aggregation features

    Returns:
        df : DataFrame with interaction features added
    """
    print("  ► Creating interaction features...")

    # Ratio of current transaction to card's historical average
    # Handles division by zero with +1e-6 epsilon
    df['amt_per_card_avg'] = df['TransactionAmt'] / (df['card_amt_mean'] + 1e-6)

    # Time interaction: night transactions on weekends are more suspicious
    df['night_x_weekend'] = df['is_night'] * df['is_weekend']

    # Log-transform of transaction amount (reduces skewness)
    # log(1 + x) avoids log(0) for zero-value transactions
    df['log_transaction_amt'] = np.log1p(df['TransactionAmt'])

    print(f"    Added 3 interaction features")
    return df


# ────────────────────────────────────────────────────────────
def save_feature_summary(df: pd.DataFrame, target_col: str, output_path: str):
    """
    Save feature-level summary with correlation to fraud label.

    Computes absolute Pearson correlation between each feature
    and the isFraud target as a proxy for individual feature importance.

    Args:
        df         : feature-engineered DataFrame
        target_col : name of the target column ('isFraud')
        output_path: path to save summary CSV
    """
    print("  ► Saving feature summary...")

    num_cols = df.select_dtypes(include=np.number).columns.tolist()
    num_cols = [c for c in num_cols if c != 'TransactionID']

    # Absolute correlation with target (fraud)
    corr_with_target = df[num_cols].corrwith(df[target_col]).abs()

    summary = pd.DataFrame({
        'feature'          : corr_with_target.index,
        'corr_with_fraud'  : corr_with_target.values,
        'dtype'            : df[corr_with_target.index].dtypes.values,
        'missing_pct'      : df[corr_with_target.index].isnull().mean().values * 100,
    }).sort_values('corr_with_fraud', ascending=False)

    summary.to_csv(output_path, index=False)
    print(f"  ✓ Feature summary saved → {output_path}")
    print(f"    Top 5 features by fraud correlation:")
    print(summary.head(5)[['feature', 'corr_with_fraud']].to_string(index=False))


# ────────────────────────────────────────────────────────────
def run_module3(df: pd.DataFrame = None) -> pd.DataFrame:
    """
    Orchestrates Module 3 feature engineering pipeline.

    Returns:
        df : DataFrame with all engineered features
    """
    if df is None:
        print(f"\n[MODULE 3] Loading from {INPUT_DATA}...")
        df = pd.read_csv(INPUT_DATA)

    print("\n[MODULE 3] ── Feature Engineering ──")
    initial_cols = df.shape[1]

    df = create_temporal_features(df)
    df = create_card_aggregation_features(df)
    df = create_email_features(df)
    df = create_interaction_features(df)

    print(f"\n  ✓ Feature count: {initial_cols} → {df.shape[1]} (+{df.shape[1]-initial_cols} new)")

    save_feature_summary(df, 'isFraud', OUTPUT_SUMMARY)

    df.to_csv(OUTPUT_FEATURES, index=False)
    print(f"  ✓ Engineered features saved → {OUTPUT_FEATURES}")
    print("[MODULE 3] Complete ✓\n")

    return df


# ── Execute ──────────────────────────────────────────────────
if __name__ == "__main__":
    df_feat = run_module3()
```

⚙️ Module 4: Graph Construction (Spatial Fraud Network)

```python
# ============================================================
# MODULE 4: FRAUD GRAPH CONSTRUCTION
# ============================================================
# PURPOSE  : Build a transaction graph where:
#              - Nodes  = unique transactions (with feature vectors)
#              - Edges  = two transactions share the same card1
#                         OR same P_emaildomain OR same addr1
#            This graph captures fraud ring structure:
#            fraudsters reuse cards/emails/addresses.
# INPUT    : module3_engineered_features.csv
# OUTPUT   : module4_node_features.csv  (per-transaction feature matrix)
#            module4_edges.csv          (edge list: src, dst, edge_type)
#            module4_graph_stats.csv    (graph statistics summary)
# ============================================================

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

INPUT_FEATURES  = 'module3_engineered_features.csv'
OUTPUT_NODES    = 'module4_node_features.csv'
OUTPUT_EDGES    = 'module4_edges.csv'
OUTPUT_STATS    = 'module4_graph_stats.csv'

# Columns to use as graph connection keys (shared identifiers)
EDGE_KEYS = ['card1', 'P_emaildomain', 'addr1']

# Columns to use as node features for GNN
NODE_FEATURE_COLS = [
    'TransactionAmt', 'log_transaction_amt',
    'card_tx_count', 'card_amt_mean', 'card_amt_std',
    'card_fraud_rate', 'email_fraud_rate',
    'hour_of_day', 'day_of_week', 'is_weekend', 'is_night',
    'sin_hour', 'cos_hour', 'sin_day', 'cos_day',
    'amt_per_card_avg', 'night_x_weekend',
]

# Max edges to build per key to prevent memory explosion
# IEEE CIS has 590K transactions; shared card1 could create millions of edges
MAX_EDGES_PER_GROUP = 50


# ────────────────────────────────────────────────────────────
def build_edges_from_shared_key(df: pd.DataFrame,
                                 key: str,
                                 max_per_group: int) -> pd.DataFrame:
    """
    Create edges between transactions that share the same value in 'key'.

    For each unique value of 'key', connect all transaction pairs
    in that group. E.g., key='card1' → all transactions from
    the same card form a clique (fully connected subgraph).

    To control graph size, limit each group to max_per_group edges
    by sampling if the group is too large.

    Args:
        df            : DataFrame with TransactionID and key column
        key           : column name for grouping (e.g., 'card1')
        max_per_group : max edges per group to prevent graph explosion

    Returns:
        edges_df : DataFrame with columns [src, dst, edge_type]
                   src and dst are integer row indices (0-based)
    """
    print(f"    Building edges from shared '{key}'...")

    edges = []                     # list of (src_idx, dst_idx) tuples
    groups = df.groupby(key)       # group transactions by shared key value

    for group_val, group_df in groups:
        indices = group_df.index.tolist()  # row indices of transactions in group

        if len(indices) < 2:
            continue  # Single transaction → no edges possible

        # Generate all pairs within the group (creates clique)
        # For large groups, sample random pairs to limit edge count
        pairs = []
        for i in range(len(indices)):
            for j in range(i + 1, len(indices)):
                pairs.append((indices[i], indices[j]))
                if len(pairs) >= max_per_group:
                    break
            if len(pairs) >= max_per_group:
                break

        edges.extend(pairs)

    # Convert to DataFrame; each edge appears in both directions (undirected)
    if not edges:
        return pd.DataFrame(columns=['src', 'dst', 'edge_type'])

    src_list = [e[0] for e in edges] + [e[1] for e in edges]
    dst_list = [e[1] for e in edges] + [e[0] for e in edges]

    edges_df = pd.DataFrame({
        'src'      : src_list,           # source transaction index
        'dst'      : dst_list,           # destination transaction index
        'edge_type': key,                # which key created this edge
    })

    print(f"      Created {len(edges_df):,} edges (undirected) from '{key}'")
    return edges_df


# ────────────────────────────────────────────────────────────
def build_graph(df: pd.DataFrame) -> tuple:
    """
    Build the full fraud transaction graph by combining edges
    from all EDGE_KEYS (card1, P_emaildomain, addr1).

    The resulting graph:
      - Nodes: each transaction is a node
      - Edges: two transactions are connected if they share
               card1 OR email domain OR billing address

    Args:
        df : feature-engineered DataFrame

    Returns:
        node_features : DataFrame of node feature matrix
        all_edges     : DataFrame with [src, dst, edge_type]
    """
    print("\n  ► Building fraud transaction graph...")
    print(f"    Nodes (transactions): {len(df):,}")
    print(f"    Edge keys: {EDGE_KEYS}")

    # Reset index to ensure 0-based integer indexing for graph edges
    df = df.reset_index(drop=True)

    # ── Build edges from each shared key ────────────────────
    all_edges_list = []
    for key in EDGE_KEYS:
        if key in df.columns:
            edges_df = build_edges_from_shared_key(df, key, MAX_EDGES_PER_GROUP)
            all_edges_list.append(edges_df)

    # Combine all edges and remove duplicates (same pair from different keys)
    all_edges = pd.concat(all_edges_list, ignore_index=True)
    all_edges = all_edges.drop_duplicates(subset=['src', 'dst'])
    print(f"\n  ► Total unique edges: {len(all_edges):,}")

    # ── Extract node features ────────────────────────────────
    # Only use features that exist in the dataframe
    available_node_cols = [c for c in NODE_FEATURE_COLS if c in df.columns]
    node_features = df[available_node_cols + ['TransactionID', 'isFraud']].copy()

    print(f"  ► Node feature dimensions: {len(available_node_cols)}")

    return node_features, all_edges


# ────────────────────────────────────────────────────────────
def compute_graph_statistics(node_features: pd.DataFrame,
                              edges: pd.DataFrame) -> pd.DataFrame:
    """
    Compute and report key graph statistics.

    Args:
        node_features : node feature DataFrame
        edges         : edge list DataFrame

    Returns:
        stats_df : DataFrame with graph statistics
    """
    n_nodes  = len(node_features)
    n_edges  = len(edges)

    # Degree = number of edges per node
    degree = edges.groupby('src')['dst'].count()

    # Density = actual edges / maximum possible edges
    max_edges = n_nodes * (n_nodes - 1)
    density   = n_edges / max_edges if max_edges > 0 else 0

    # Fraud node ratio
    fraud_ratio = node_features['isFraud'].mean()

    stats = {
        'n_nodes'             : n_nodes,
        'n_edges'             : n_edges,
        'graph_density'       : round(density, 8),
        'avg_node_degree'     : round(degree.mean(), 2),
        'max_node_degree'     : int(degree.max()),
        'fraud_node_ratio'    : round(fraud_ratio, 4),
        'isolated_nodes'      : n_nodes - degree.shape[0],  # nodes with no edges
    }

    print("\n  ── Graph Statistics ──────────────────────────")
    for k, v in stats.items():
        print(f"    {k:<25} : {v}")

    return pd.DataFrame([stats])


# ────────────────────────────────────────────────────────────
def run_module4(df: pd.DataFrame = None) -> tuple:
    """
    Orchestrates Module 4: graph construction and saving.

    Returns:
        node_features : DataFrame with node feature matrix
        edges         : DataFrame with edge list
    """
    if df is None:
        print(f"\n[MODULE 4] Loading from {INPUT_FEATURES}...")
        df = pd.read_csv(INPUT_FEATURES)

    print("\n[MODULE 4] ── Graph Construction ──")

    node_features, edges = build_graph(df)
    stats_df = compute_graph_statistics(node_features, edges)

    # Save all outputs
    node_features.to_csv(OUTPUT_NODES, index=True)
    edges.to_csv(OUTPUT_EDGES, index=False)
    stats_df.to_csv(OUTPUT_STATS, index=False)

    print(f"\n  ✓ Node features saved   → {OUTPUT_NODES}")
    print(f"  ✓ Edge list saved       → {OUTPUT_EDGES}")
    print(f"  ✓ Graph stats saved     → {OUTPUT_STATS}")
    print("[MODULE 4] Complete ✓\n")

    return node_features, edges


# ── Execute ──────────────────────────────────────────────────
if __name__ == "__main__":
    nodes, edges = run_module4()
```

⚙️ Module 5: Model Architectures (3 Models + SpaTempFraudNet)

```python
# ============================================================
# MODULE 5: DEEP LEARNING MODEL ARCHITECTURES
# ============================================================
# PURPOSE  : Define three model architectures:
#   [Model A] AdaptiveTransformer  — tabular feature learning
#   [Model B] GraphProposalNet     — graph-based spatial learning
#   [Model C] SpaTempFraudNet      — FUSION (thesis model)
#
# Mathematical Foundations:
#   Adaptive Attention:
#     m_h(x) = clamp((s_h + R/2 - x)/R, 0, 1)
#     Attn_h = softmax(Q_h K_h^T / sqrt(d) * m_h) V_h
#
#   Graph Attention:
#     α_ij = softmax(LeakyReLU(a^T [Wh_i || Wh_j]))
#     h'_i = σ(Σ_{j∈N(i)} α_ij W h_j)
#
#   Temporal Encoding:
#     PE(t, 2i)   = sin(t / 10000^(2i/d))
#     PE(t, 2i+1) = cos(t / 10000^(2i/d))
#
#   Focal Loss (for class imbalance):
#     L = -α(1-p)^γ log(p),  γ=2
# ============================================================

import torch
import torch.nn as nn
import torch.nn.functional as F
import math

# ────────────────────────────────────────────────────────────
# ── COMPONENT 1: Temporal Positional Encoding ───────────────
# ────────────────────────────────────────────────────────────

class TemporalPositionalEncoding(nn.Module):
    """
    Sinusoidal positional encoding adapted for transaction time features.

    Standard transformer PE encodes position in a sequence.
    Here we encode a transaction's temporal features (hour, day)
    as a continuous positional representation.

    Formulas:
        PE(t, 2i)   = sin(t / 10000^(2i / d_model))
        PE(t, 2i+1) = cos(t / 10000^(2i / d_model))

    This creates a d_model-dimensional time embedding that is:
      - Deterministic (no learned parameters)
      - Smooth (nearby times get similar encodings)
      - Periodic (captures daily/weekly cycles)

    Args:
        d_model : embedding dimension (must match transformer d_model)
        max_len : maximum temporal index (10000 for seconds in a day)
    """
    def __init__(self, d_model: int, max_len: int = 10000):
        super().__init__()

        # Create a (max_len, d_model) PE matrix
        pe = torch.zeros(max_len, d_model)                            # (T, D)

        # Position indices: [0, 1, 2, ..., max_len-1] as column vector
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)  # (T, 1)

        # Denominator term: 10000^(2i/d_model) for each dimension pair i
        div_term = torch.exp(
            torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model)
        )  # (D/2,)

        # Apply sin to even indices, cos to odd indices
        pe[:, 0::2] = torch.sin(position * div_term)   # even dims: sin
        pe[:, 1::2] = torch.cos(position * div_term)   # odd dims: cos

        # Register as buffer (not a parameter → not updated during backprop)
        self.register_buffer('pe', pe)

    def forward(self, time_indices: torch.Tensor) -> torch.Tensor:
        """
        Look up temporal encoding for given time indices.

        Args:
            time_indices : integer tensor of time values (batch_size,)
                           e.g., hour_of_day (0-23) or TransactionDT

        Returns:
            Tensor of shape (batch_size, d_model) with time embeddings
        """
        # Clamp indices to max_len to avoid out-of-bounds
        time_indices = time_indices.long().clamp(0, self.pe.size(0) - 1)
        return self.pe[time_indices]  # (batch_size, d_model)


# ────────────────────────────────────────────────────────────
# ── COMPONENT 2: Adaptive Span Attention Head ───────────────
# ────────────────────────────────────────────────────────────

class AdaptiveSpanAttention(nn.Module):
    """
    Multi-head attention where each head learns its own attention span.

    Standard MHA treats all features symmetrically. In fraud detection,
    different feature groups (card features, V-features, time features)
    have different correlation ranges. Adaptive span allows Head 1 to
    attend to nearby features only, while Head 2 attends globally.

    Span masking formula:
        m_h(x) = clamp((s_h + R/2 - x) / R, 0, 1)

    where:
        s_h : learnable span parameter for head h
        R   : ramp size (controls softness of the cutoff)
        x   : distance between feature positions

    The mask is elementwise multiplied with attention scores,
    effectively zeroing out attention beyond the learned span.

    Args:
        d_model    : model dimension (must be divisible by n_heads)
        n_heads    : number of attention heads
        max_span   : maximum allowed attention span
        ramp_size  : ramp size R for soft masking
        dropout    : attention dropout rate
    """
    def __init__(self, d_model: int, n_heads: int,
                 max_span: int = 50, ramp_size: int = 4, dropout: float = 0.1):
        super().__init__()

        assert d_model % n_heads == 0, "d_model must be divisible by n_heads"

        self.d_model   = d_model
        self.n_heads   = n_heads
        self.d_head    = d_model // n_heads   # dimension per head
        self.max_span  = max_span
        self.ramp_size = ramp_size

        # Linear projections for Query, Key, Value and Output
        self.W_q = nn.Linear(d_model, d_model, bias=False)  # Query projection
        self.W_k = nn.Linear(d_model, d_model, bias=False)  # Key projection
        self.W_v = nn.Linear(d_model, d_model, bias=False)  # Value projection
        self.W_o = nn.Linear(d_model, d_model)               # Output projection

        # Learnable span per head: initialized to half of max_span
        # Sigmoid is applied later to keep spans in valid range
        self.span_params = nn.Parameter(
            torch.ones(n_heads) * 0.5   # initial value before sigmoid scaling
        )

        self.dropout = nn.Dropout(dropout)
        self.scale   = self.d_head ** -0.5  # scaling factor: 1/sqrt(d_head)

    def get_spans(self) -> torch.Tensor:
        """
        Compute actual span values from learnable parameters.
        Uses sigmoid to map to (0, max_span).

        Returns:
            spans : (n_heads,) tensor with span values in [0, max_span]
        """
        return torch.sigmoid(self.span_params) * self.max_span

    def compute_span_mask(self, seq_len: int, device: torch.device) -> torch.Tensor:
        """
        Compute soft position-based attention mask for each head.

        For each head h and feature pair (i,j):
            mask[h, i, j] = clamp((s_h + R/2 - |i-j|) / R, 0, 1)

        Returns:
            mask : (n_heads, seq_len, seq_len) float tensor in [0, 1]
        """
        spans = self.get_spans()  # (n_heads,)
        R     = self.ramp_size

        # Compute pairwise distances between feature positions
        pos  = torch.arange(seq_len, device=device).float()  # (S,)
        dist = (pos.unsqueeze(0) - pos.unsqueeze(1)).abs()   # (S, S)

        # For each head, compute soft mask
        # spans: (n_heads,) → (n_heads, 1, 1) for broadcasting
        s    = spans.view(self.n_heads, 1, 1)
        mask = ((s + R / 2 - dist.unsqueeze(0)) / R).clamp(0, 1)  # (H, S, S)

        return mask

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Forward pass: adaptive-span multi-head attention.

        Args:
            x : (batch_size, seq_len, d_model) input tensor
                seq_len = number of input features (after embedding)

        Returns:
            out : (batch_size, seq_len, d_model) attended features
        """
        B, S, D = x.shape  # batch, seq_len (features), d_model

        # ── Project to Q, K, V ──────────────────────────────
        # Linear projection then reshape into multi-head format
        # (B, S, D) → (B, S, H, d_head) → (B, H, S, d_head)
        Q = self.W_q(x).view(B, S, self.n_heads, self.d_head).permute(0, 2, 1, 3)
        K = self.W_k(x).view(B, S, self.n_heads, self.d_head).permute(0, 2, 1, 3)
        V = self.W_v(x).view(B, S, self.n_heads, self.d_head).permute(0, 2, 1, 3)

        # ── Scaled dot-product attention scores ─────────────
        # scores[b, h, i, j] = (Q[b,h,i] · K[b,h,j]) / sqrt(d_head)
        scores = torch.matmul(Q, K.transpose(-2, -1)) * self.scale  # (B, H, S, S)

        # ── Apply adaptive span mask ─────────────────────────
        # mask: (H, S, S) → (1, H, S, S) for batch broadcasting
        mask   = self.compute_span_mask(S, x.device).unsqueeze(0)  # (1, H, S, S)

        # Multiply scores by mask; blocked positions get large negative value
        scores = scores * mask + (1.0 - mask) * (-1e9)

        # ── Softmax → attention weights → dropout ────────────
        attn_w = F.softmax(scores, dim=-1)    # (B, H, S, S)
        attn_w = self.dropout(attn_w)

        # ── Weighted sum of values ───────────────────────────
        # context: (B, H, S, d_head) → concat heads → (B, S, D)
        context = torch.matmul(attn_w, V)                        # (B, H, S, d_head)
        context = context.permute(0, 2, 1, 3).contiguous()       # (B, S, H, d_head)
        context = context.view(B, S, D)                          # (B, S, D)

        return self.W_o(context)  # final linear projection


# ────────────────────────────────────────────────────────────
# ── MODEL A: Adaptive Transformer for Tabular Features ──────
# ────────────────────────────────────────────────────────────

class AdaptiveTransformer(nn.Module):
    """
    Custom Adaptive Transformer for tabular fraud features.

    Architecture:
        Input (batch, n_features)
        → Feature Embedding (linear: n_features → d_model, then reshape)
        → [N × Adaptive Transformer Encoder Blocks]
           Each block: AdaptiveSpanAttention → LayerNorm → FFN → LayerNorm
        → Global average pooling across feature dimension
        → Output embedding (batch, d_model)

    The output is a d_model-dimensional representation of each transaction,
    encoding complex feature interactions discovered by adaptive attention.

    Args:
        n_features  : number of input tabular features
        d_model     : transformer hidden dimension (default: 128)
        n_heads     : number of attention heads (default: 8)
        n_layers    : number of transformer encoder layers (default: 3)
        ffn_dim     : feed-forward network hidden dim (default: 256)
        dropout     : dropout rate (default: 0.1)
        max_span    : maximum adaptive attention span (default: 50)
    """
    def __init__(self, n_features: int, d_model: int = 128,
                 n_heads: int = 8, n_layers: int = 3,
                 ffn_dim: int = 256, dropout: float = 0.1,
                 max_span: int = 50):
        super().__init__()

        self.d_model = d_model

        # ── Feature embedding ────────────────────────────────
        # Projects each scalar feature to a d_model-dimensional vector
        # (batch, n_features) → (batch, n_features, d_model)
        self.feature_embedding = nn.Sequential(
            nn.Linear(1, d_model),          # each feature scalar → embedding
            nn.LayerNorm(d_model),
            nn.GELU(),
        )

        # ── Stack of adaptive transformer encoder layers ─────
        self.layers = nn.ModuleList([
            AdaptiveTransformerLayer(d_model, n_heads, ffn_dim, dropout, max_span)
            for _ in range(n_layers)         # n_layers identical blocks
        ])

        # ── Final layer norm ─────────────────────────────────
        self.layer_norm = nn.LayerNorm(d_model)

        # ── Output projection ────────────────────────────────
        # After pooling: (batch, d_model) → (batch, d_model)
        self.output_proj = nn.Linear(d_model, d_model)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Forward pass through the Adaptive Transformer.

        Args:
            x : (batch_size, n_features) tabular feature tensor

        Returns:
            h : (batch_size, d_model) transaction embedding
        """
        B, F = x.shape   # batch size, number of features

        # ── Embed each feature independently ─────────────────
        # (B, F) → (B, F, 1) → (B, F, d_model)
        x = x.unsqueeze(-1)             # add feature dim: (B, F, 1)
        x = self.feature_embedding(x)   # embed each feature: (B, F, d_model)

        # ── Pass through transformer layers ──────────────────
        for layer in self.layers:
            x = layer(x)   # each layer: (B, F, d_model) → (B, F, d_model)

        x = self.layer_norm(x)  # final normalization

        # ── Global average pooling across feature dimension ──
        # Aggregates information from all features into one vector
        h = x.mean(dim=1)   # (B, F, d_model) → (B, d_model)

        return self.output_proj(h)  # (B, d_model)


class AdaptiveTransformerLayer(nn.Module):
    """
    Single Adaptive Transformer encoder layer.

    Structure:
        x → AdaptiveSpanAttention → x + residual → LayerNorm
          → FFN (Linear → GELU → Dropout → Linear) → x + residual → LayerNorm

    Args:
        d_model  : model dimension
        n_heads  : attention heads
        ffn_dim  : feed-forward hidden dimension
        dropout  : dropout rate
        max_span : maximum adaptive attention span
    """
    def __init__(self, d_model, n_heads, ffn_dim, dropout, max_span):
        super().__init__()

        # Adaptive span self-attention block
        self.attn      = AdaptiveSpanAttention(d_model, n_heads, max_span, dropout=dropout)
        self.norm1     = nn.LayerNorm(d_model)   # post-attention norm

        # Position-wise Feed-Forward Network (FFN)
        # Expands to ffn_dim then projects back to d_model
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim),   # expand: d_model → ffn_dim
            nn.GELU(),                      # GELU activation (smoother than ReLU)
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, d_model),   # project back: ffn_dim → d_model
            nn.Dropout(dropout),
        )
        self.norm2 = nn.LayerNorm(d_model)  # post-FFN norm

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Pre-norm transformer block with residual connections.

        Args:
            x : (batch, seq_len, d_model)

        Returns:
            x : (batch, seq_len, d_model)
        """
        # Attention block with residual connection (x + attention output)
        x = x + self.attn(self.norm1(x))

        # FFN block with residual connection
        x = x + self.ffn(self.norm2(x))

        return x


# ────────────────────────────────────────────────────────────
# ── MODEL B: Graph Proposal Neural Network (GPNN) ───────────
# ────────────────────────────────────────────────────────────

class GraphAttentionLayer(nn.Module):
    """
    Sparse multi-head graph attention.

    IMPORTANT:
    This implementation computes attention only on existing edges.

    Complexity:
        Old implementation: O(N^2 * H)
        New implementation: O(E * H)

    For IEEE-CIS:
        N ≈ 590,540
        E ≈ 551,566

    Therefore we avoid constructing an enormous N x N attention matrix.
    """

    def __init__(
        self,
        in_dim,
        out_dim,
        n_heads=4,
        dropout=0.1,
        concat=True
    ):
        super().__init__()

        self.in_dim = in_dim
        self.out_dim = out_dim
        self.n_heads = n_heads
        self.concat = concat

        self.W = nn.Linear(
            in_dim,
            out_dim * n_heads,
            bias=False
        )

        self.att_src = nn.Parameter(
            torch.empty(n_heads, out_dim)
        )

        self.att_dst = nn.Parameter(
            torch.empty(n_heads, out_dim)
        )

        self.bias = nn.Parameter(
            torch.zeros(out_dim * n_heads if concat else out_dim)
        )

        self.dropout = nn.Dropout(dropout)
        self.leaky_relu = nn.LeakyReLU(0.2)

        self.reset_parameters()

    def reset_parameters(self):

        nn.init.xavier_uniform_(self.W.weight)

        nn.init.xavier_uniform_(self.att_src)
        nn.init.xavier_uniform_(self.att_dst)

        nn.init.zeros_(self.bias)

    def forward(self, h, edge_index):
        """
        Args:
            h:
                Node features
                Shape: (N, in_dim)

            edge_index:
                Graph edges
                Shape: (2, E)

        Returns:
            Node embeddings
            Shape:
                (N, out_dim * n_heads) if concat=True
                (N, out_dim) otherwise
        """

        N = h.size(0)

        # ------------------------------------------------------
        # Linear projection
        # ------------------------------------------------------

        # (N, H * D)
        Wh = self.W(h)

        # (N, H, D)
        Wh = Wh.view(
            N,
            self.n_heads,
            self.out_dim
        )

        src = edge_index[0]
        dst = edge_index[1]

        # ------------------------------------------------------
        # Gather source/destination node representations
        # ------------------------------------------------------

        Wh_src = Wh[src]   # (E, H, D)
        Wh_dst = Wh[dst]   # (E, H, D)

        # ------------------------------------------------------
        # Attention coefficients
        # ------------------------------------------------------

        alpha_src = (
            Wh_src * self.att_src.unsqueeze(0)
        ).sum(dim=-1)

        alpha_dst = (
            Wh_dst * self.att_dst.unsqueeze(0)
        ).sum(dim=-1)

        # (E, H)
        alpha = self.leaky_relu(
            alpha_src + alpha_dst
        )

        # ------------------------------------------------------
        # Sparse softmax over incoming edges
        # ------------------------------------------------------

        # Maximum attention score per destination node/head
        max_alpha = torch.full(
            (N, self.n_heads),
            -torch.inf,
            device=h.device,
            dtype=alpha.dtype
        )

        max_alpha.scatter_reduce_(
            0,
            dst.unsqueeze(-1).expand(-1, self.n_heads),
            alpha,
            reduce="amax",
            include_self=True
        )

        # Numerical stability
        alpha_exp = torch.exp(
            alpha - max_alpha[dst]
        )

        # Sum of exponentials per destination node/head
        alpha_sum = torch.zeros(
            (N, self.n_heads),
            device=h.device,
            dtype=alpha.dtype
        )

        alpha_sum.scatter_add_(
            0,
            dst.unsqueeze(-1).expand(-1, self.n_heads),
            alpha_exp
        )

        # Avoid division by zero
        alpha = alpha_exp / (
            alpha_sum[dst] + 1e-12
        )

        alpha = self.dropout(alpha)

        # ------------------------------------------------------
        # Message aggregation
        # ------------------------------------------------------

        messages = Wh_src * alpha.unsqueeze(-1)

        out = torch.zeros(
            (N, self.n_heads, self.out_dim),
            device=h.device,
            dtype=messages.dtype
        )

        index = dst.view(-1, 1, 1).expand(
            -1,
            self.n_heads,
            self.out_dim
        )

        out.scatter_add_(
            0,
            index,
            messages
        )

        # ------------------------------------------------------
        # Combine heads
        # ------------------------------------------------------

        if self.concat:
            out = out.reshape(
                N,
                self.n_heads * self.out_dim
            )
        else:
            out = out.mean(dim=1)

        out = out + self.bias

        return out


class GraphProposalNet(nn.Module):
    """
    Graph Proposal Neural Network (GPNN) for fraud detection.

    Inspired by Region Proposal Networks (RPN) in object detection,
    but applied to transaction graphs.

    Architecture:
        Node Features (N, node_feat_dim)
        → GAT Layer 1 → node embeddings (N, H × F1)
        → GAT Layer 2 → refined embeddings (N, F2)
        → Proposal Scorer: s_i = σ(W_p h_i + b_p)
        → Top-K proposal selection (select suspicious nodes)
        → Subgraph refinement (second GAT pass on proposals)
        → Output: (N, out_dim) graph representations

    The proposal mechanism acts as an attention bottleneck that
    focuses computational resources on the most suspicious transactions.

    Args:
        node_feat_dim  : dimension of input node features
        hidden_dim     : GAT hidden dimension
        out_dim        : output embedding dimension
        n_heads        : number of GAT attention heads
        top_k_ratio    : fraction of nodes to keep as proposals (0-1)
        dropout        : dropout rate
    """
    def __init__(self, node_feat_dim: int, hidden_dim: int = 64,
                 out_dim: int = 128, n_heads: int = 4,
                 top_k_ratio: float = 0.3, dropout: float = 0.1):
        super().__init__()

        self.top_k_ratio = top_k_ratio
        self.out_dim     = out_dim

        # ── GAT Layer 1 (feature extraction) ─────────────────
        # Input: (N, node_feat_dim) → Output: (N, hidden_dim * n_heads)
        self.gat1 = GraphAttentionLayer(
            in_dim=node_feat_dim,
            out_dim=hidden_dim,
            n_heads=n_heads,
            dropout=dropout,
            concat=True   # concatenate heads
        )
        gat1_out_dim = hidden_dim * n_heads  # size after concat

        # ── Batch normalization after GAT1 ───────────────────
        self.bn1 = nn.BatchNorm1d(gat1_out_dim)

        # ── GAT Layer 2 (refinement) ─────────────────────────
        # Input: (N, hidden_dim*n_heads) → Output: (N, out_dim)
        self.gat2 = GraphAttentionLayer(
            in_dim=gat1_out_dim,
            out_dim=out_dim,
            n_heads=1,          # single head for final layer
            dropout=dropout,
            concat=False  # average heads → (N, out_dim)
        )

        # ── Proposal scoring head ─────────────────────────────
        # Assigns a suspicion score to each node: s_i = σ(Wh_i + b)
        # This is the "proposal" mechanism — nodes with high s are suspicious
        self.proposal_scorer = nn.Sequential(
            nn.Linear(gat1_out_dim, 32),   # compress
            nn.ReLU(),
            nn.Linear(32, 1),              # scalar score per node
            nn.Sigmoid(),                  # → [0, 1] suspicion score
        )

        # ── Node-level output projection ─────────────────────
        self.output_proj = nn.Sequential(
            nn.Linear(out_dim, out_dim),
            nn.LayerNorm(out_dim),
            nn.GELU(),
        )

        self.dropout = nn.Dropout(dropout)

    def forward(self, node_features: torch.Tensor,
                edge_index: torch.Tensor) -> tuple:
        """
        Forward pass through Graph Proposal Network.

        Args:
            node_features : (N, node_feat_dim) node feature matrix
            edge_index    : (2, E) edge connectivity

        Returns:
            node_emb    : (N, out_dim) node embeddings
            prop_scores : (N, 1) proposal scores (fraud suspicion)
        """
        # ── Step 1: First GAT pass (feature extraction) ──────
        h1 = self.gat1(node_features, edge_index)   # (N, H*hidden_dim)
        h1 = self.bn1(h1)                            # batch normalize
        h1 = F.gelu(h1)                              # non-linear activation
        h1 = self.dropout(h1)

        # ── Step 2: Compute proposal scores ──────────────────
        # Each node gets a fraud suspicion score: s_i ∈ [0, 1]
        prop_scores = self.proposal_scorer(h1)  # (N, 1)

        # ── Step 3: Top-K proposal selection ─────────────────
        # Select top-K most suspicious nodes for focused refinement
        # This implements the "proposal" mechanism
        N = node_features.shape[0]
        K = max(1, int(N * self.top_k_ratio))          # minimum 1 proposal

        # Get indices of top-K scoring nodes
        _, top_k_indices = torch.topk(
            prop_scores.squeeze(-1),   # (N,) scores
            k=min(K, N),               # can't select more than N nodes
            largest=True,              # highest scores = most suspicious
        )

        # ── Step 4: Second GAT pass (refinement) ─────────────
        # Run second GAT on full graph, but proposal scores guide attention
        h2 = self.gat2(h1, edge_index)  # (N, out_dim)
        h2 = F.gelu(h2)

        # Weight node embeddings by proposal scores
        # High-suspicion nodes get amplified; low-suspicion nodes attenuated
        h2 = h2 * prop_scores  # element-wise scaling: (N, out_dim)

        # ── Step 5: Output projection ─────────────────────────
        node_emb = self.output_proj(h2)  # (N, out_dim)

        return node_emb, prop_scores


# ────────────────────────────────────────────────────────────
# ── FOCAL LOSS (for class imbalance) ────────────────────────
# ────────────────────────────────────────────────────────────

class FocalLoss(nn.Module):
    """
    Focal Loss for binary classification with severe class imbalance.

    IEEE CIS dataset has ~3.5% fraud rate.
    Standard BCE loss is dominated by easy negatives (96.5% legit).
    Focal loss down-weights easy examples, focusing training on hard ones.

    Formula:
        FL(p_t) = -α_t (1 - p_t)^γ log(p_t)

    where:
        p_t = predicted probability for the true class
        γ   = focusing parameter (default: 2)
              γ=0 → standard BCE; γ>0 → focus on hard examples
        α_t = class weight (0.25 for negatives, 0.75 for positives)

    Args:
        gamma : focusing parameter (default: 2)
        alpha : weight for positive class (default: 0.75 for rare fraud)
    """
    def __init__(self, gamma: float = 2.0, alpha: float = 0.75):
        super().__init__()
        self.gamma = gamma
        self.alpha = alpha

    def forward(self, logits: torch.Tensor,
                targets: torch.Tensor) -> torch.Tensor:
        """
        Compute focal loss.

        Args:
            logits  : (batch_size,) raw model outputs (before sigmoid)
            targets : (batch_size,) binary labels {0, 1}

        Returns:
            loss : scalar focal loss value
        """
        # Convert logits to probabilities
        probs = torch.sigmoid(logits)              # p ∈ [0, 1]

        # Probability for the true class:
        # p_t = p if y=1, else (1-p)
        p_t = probs * targets + (1 - probs) * (1 - targets)

        # Class weight:
        # α_t = α if y=1, else (1-α)
        alpha_t = self.alpha * targets + (1 - self.alpha) * (1 - targets)

        # Focal weight: (1 - p_t)^γ
        # Easy examples (p_t → 1) → small weight
        # Hard examples (p_t → 0) → weight → 1
        focal_weight = (1 - p_t) ** self.gamma

        # Compute binary cross-entropy component
        bce = F.binary_cross_entropy_with_logits(logits, targets, reduction='none')

        # Combine: FL = α_t * (1-p_t)^γ * BCE
        focal_loss = alpha_t * focal_weight * bce

        return focal_loss.mean()


# ────────────────────────────────────────────────────────────
# ── MODEL C: SpaTempFraudNet (MAIN THESIS MODEL) ────────────
# ────────────────────────────────────────────────────────────

class SpaTempFraudNet(nn.Module):
    """
    Spatio-Temporal Fraud Detection Network — Thesis Contribution.

    Combines three complementary representations:
      1. Adaptive Transformer (ATF)  → tabular feature interactions
      2. Graph Proposal Net (GPNN)   → spatial fraud ring structure
      3. Temporal Encoding (TPE)     → time-based fraud patterns

    Fusion:
        h = MLP([h_ATF || h_GPNN || h_TPE])
        p(fraud) = σ(W_out h + b_out)

    The model learns to weight each modality through the fusion MLP,
    allowing it to prioritize graph signals when fraud rings exist
    and temporal signals when time-based patterns dominate.

    Args:
        n_tabular_features : number of tabular input features
        node_feat_dim      : node feature dimension for GNN
        d_model            : ATF/fusion model dimension (default: 128)
        gnn_hidden         : GNN hidden dimension (default: 64)
        gnn_out_dim        : GNN output dimension (default: 128)
        n_atf_heads        : ATF attention heads (default: 8)
        n_atf_layers       : ATF encoder layers (default: 3)
        dropout            : dropout rate (default: 0.1)
        max_span           : ATF maximum attention span (default: 50)
    """
    def __init__(self,
                 n_tabular_features: int,
                 node_feat_dim: int,
                 d_model: int = 128,
                 gnn_hidden: int = 64,
                 gnn_out_dim: int = 128,
                 n_atf_heads: int = 8,
                 n_atf_layers: int = 3,
                 dropout: float = 0.1,
                 max_span: int = 50):
        super().__init__()

        self.d_model = d_model

        # ── Module 1: Adaptive Transformer ───────────────────
        # Processes tabular features (V-features, C-features, etc.)
        self.adaptive_transformer = AdaptiveTransformer(
            n_features=n_tabular_features,
            d_model=d_model,
            n_heads=n_atf_heads,
            n_layers=n_atf_layers,
            ffn_dim=d_model * 2,
            dropout=dropout,
            max_span=max_span,
        )

        # ── Module 2: Graph Proposal Net ─────────────────────
        # Processes graph-structured spatial features
        self.graph_proposal_net = GraphProposalNet(
            node_feat_dim=node_feat_dim,
            hidden_dim=gnn_hidden,
            out_dim=gnn_out_dim,
            n_heads=4,
            top_k_ratio=0.3,
            dropout=dropout,
        )

        # ── Module 3: Temporal Positional Encoding ────────────
        # Encodes transaction time as a dense representation
        self.temporal_encoding = TemporalPositionalEncoding(d_model)

        # Learnable weight to balance temporal encoding influence
        self.temporal_gate = nn.Parameter(torch.ones(1) * 0.5)

        # ── Fusion Layer ──────────────────────────────────────
        # Combines ATF + GPNN + Temporal representations
        fusion_input_dim = d_model + gnn_out_dim + d_model  # concatenated dim

        self.fusion_mlp = nn.Sequential(
            nn.Linear(fusion_input_dim, 256),   # compress to 256
            nn.LayerNorm(256),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(256, 128),                # further compress to 128
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Dropout(dropout),
        )

        # ── Classification Head ───────────────────────────────
        # Final binary fraud/non-fraud prediction
        self.classifier = nn.Sequential(
            nn.Linear(128, 32),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(32, 1),  # single output logit (use sigmoid externally)
        )

        # ── Loss Function ─────────────────────────────────────
        self.focal_loss = FocalLoss(gamma=2.0, alpha=0.75)

        # ── Initialize weights ────────────────────────────────
        self._init_weights()

    def _init_weights(self):
        """
        Xavier uniform initialization for all linear layers.
        Helps avoid vanishing/exploding gradients at start of training.
        """
        for module in self.modules():
            if isinstance(module, nn.Linear):
                nn.init.xavier_uniform_(module.weight)
                if module.bias is not None:
                    nn.init.zeros_(module.bias)

    def forward(self,
                tabular_features: torch.Tensor,
                node_features: torch.Tensor,
                edge_index: torch.Tensor,
                time_indices: torch.Tensor) -> dict:
        """
        Forward pass through SpaTempFraudNet.

        Args:
            tabular_features : (B, n_tabular_features) batch of feature vectors
            node_features    : (N, node_feat_dim) ALL node features in the graph
                               (N can be different from B; graph may be full batch)
            edge_index       : (2, E) graph edge index
            time_indices     : (B,) integer time values (e.g., hour_of_day)

        Returns:
            dict with:
                'logits'       : (B,) raw prediction logits
                'probs'        : (B,) fraud probabilities after sigmoid
                'prop_scores'  : (N, 1) proposal scores from GPNN
                'atf_embed'    : (B, d_model) ATF embeddings
                'graph_embed'  : (N, gnn_out_dim) graph node embeddings
        """
        # ── Branch 1: Adaptive Transformer ───────────────────
        # Captures complex feature interactions in tabular data
        h_atf = self.adaptive_transformer(tabular_features)  # (B, d_model)

        # ── Branch 2: Graph Proposal Net ─────────────────────
        # Captures spatial fraud patterns in the transaction graph
        h_graph, prop_scores = self.graph_proposal_net(
            node_features, edge_index
        )  # h_graph: (N, gnn_out_dim); prop_scores: (N, 1)

        # Align graph embeddings to batch size
        # For mini-batch training: h_graph may be batch-aligned
        # Here we use the first B nodes (assumes batch-aligned graph)
        B = tabular_features.shape[0]
        h_graph_batch = h_graph[:B]  # (B, gnn_out_dim)

        # ── Branch 3: Temporal Encoding ───────────────────────
        # Captures time-based fraud patterns
        h_temp = self.temporal_encoding(time_indices)  # (B, d_model)
        h_temp = h_temp * torch.sigmoid(self.temporal_gate)  # gated scaling

        # ── Fusion: Concatenate all representations ───────────
        h_fused = torch.cat([h_atf, h_graph_batch, h_temp], dim=-1)  # (B, D_fusion)

        # ── MLP Fusion ────────────────────────────────────────
        h_fused = self.fusion_mlp(h_fused)   # (B, 128)

        # ── Classification ────────────────────────────────────
        logits = self.classifier(h_fused).squeeze(-1)   # (B,) raw logits
        probs  = torch.sigmoid(logits)                   # (B,) probabilities

        return {
            'logits'      : logits,
            'probs'       : probs,
            'prop_scores' : prop_scores,
            'atf_embed'   : h_atf,
            'graph_embed' : h_graph,
        }

    def compute_loss(self, outputs: dict, targets: torch.Tensor) -> torch.Tensor:
        """
        Compute combined loss:
            L = L_focal(main) + λ * L_proposal(auxiliary)

        The proposal loss provides a secondary supervision signal
        for the GPNN proposal scorer, improving its calibration.

        Args:
            outputs : dict from forward() containing 'logits', 'prop_scores'
            targets : (B,) binary fraud labels

        Returns:
            total_loss : scalar loss tensor
        """
        B = targets.shape[0]

        # ── Main focal loss ───────────────────────────────────
        main_loss = self.focal_loss(outputs['logits'], targets.float())

        # ── Auxiliary proposal loss ───────────────────────────
        # Use proposal scores for first B nodes as auxiliary predictions
        prop_scores = outputs['prop_scores'][:B].squeeze(-1)  # (B,)
        prop_loss   = F.binary_cross_entropy(
            prop_scores.clamp(1e-6, 1 - 1e-6),
            targets.float()
        )

        # Combined loss: main + 0.1 * auxiliary
        total_loss = main_loss + 0.1 * prop_loss

        return total_loss


# ── Model Summary Utility ────────────────────────────────────
def count_parameters(model: nn.Module) -> int:
    """Count total trainable parameters in a model."""
    return sum(p.numel() for p in model.parameters() if p.requires_grad)


# ── Quick instantiation test ──────────────────────────────────
if __name__ == "__main__":
    print("\n[MODULE 5] Testing model architectures...")

    # Hyperparameters (adjust based on actual feature count after modules 1-4)
    N_TABULAR = 80       # example: 80 tabular features after preprocessing
    N_NODES   = 200      # example: 200 nodes in mini-batch graph
    BATCH     = 64       # batch size

    # Instantiate SpaTempFraudNet
    model = SpaTempFraudNet(
        n_tabular_features=N_TABULAR,
        node_feat_dim=17,              # as defined in MODULE 4 NODE_FEATURE_COLS
        d_model=128,
        gnn_hidden=64,
        gnn_out_dim=128,
        n_atf_heads=8,
        n_atf_layers=3,
    )

    print(f"  SpaTempFraudNet parameters: {count_parameters(model):,}")

    # Dummy forward pass
    dummy_tabular = torch.randn(BATCH, N_TABULAR)
    dummy_nodes   = torch.randn(N_NODES, 17)
    dummy_edges   = torch.randint(0, N_NODES, (2, 500))
    dummy_time    = torch.randint(0, 24, (BATCH,))
    dummy_labels  = torch.randint(0, 2, (BATCH,)).float()

    with torch.no_grad():
        out  = model(dummy_tabular, dummy_nodes, dummy_edges, dummy_time)
        loss = model.compute_loss(out, dummy_labels)

    print(f"  Output logits shape : {out['logits'].shape}")
    print(f"  Output probs shape  : {out['probs'].shape}")
    print(f"  Loss value          : {loss.item():.4f}")
    print("[MODULE 5] Complete ✓\n")
```

⚙️ Module 6: Training Pipeline

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# ── Import model from Module 5 ────────────────────────────────
# SpaTempFraudNet, count_parameters are defined in a previous cell (Module 5)
# and should be available in the global scope after its execution.
# (In practice, import from your module5_models.py file)
# The following line is removed as the objects are expected to be in global scope after Module 5 execution.
# from __main__ import SpaTempFraudNet, count_parameters

# ── Configuration ────────────────────────────────────────────
TABULAR_PATH  = 'module3_engineered_features.csv'
NODE_PATH     = 'module4_node_features.csv'
EDGE_PATH     = 'module4_edges.csv'
OUTPUT_LOG    = 'module6_training_log.csv'
MODEL_SAVE    = 'module6_best_model.pt'

# Training hyperparameters
BATCH_SIZE    = 128        # mini-batch size
EPOCHS        = 30         # maximum training epochs
LR            = 1e-3       # initial learning rate
WEIGHT_DECAY  = 1e-5       # L2 regularization coefficient
PATIENCE      = 7          # early stopping patience (epochs without improvement)
VAL_SPLIT     = 0.15       # 15% for validation
TEST_SPLIT    = 0.15       # 15% for test
DEVICE        = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Feature columns for tabular model input
TABULAR_EXCLUDE = ['TransactionID', 'isFraud', 'TransactionDT']

# Time column for temporal encoding
TIME_COL = 'hour_of_day'

# Node feature columns (must match MODULE 4 NODE_FEATURE_COLS)
NODE_FEATURE_COLS = [
    'TransactionAmt', 'log_transaction_amt',
    'card_tx_count', 'card_amt_mean', 'card_amt_std',
    'card_fraud_rate', 'email_fraud_rate',
    'hour_of_day', 'day_of_week', 'is_weekend', 'is_night',
    'sin_hour', 'cos_hour', 'sin_day', 'cos_day',
    'amt_per_card_avg', 'night_x_weekend',
]


# ────────────────────────────────────────────────────────────
class FraudDataset(Dataset):
    """
    PyTorch Dataset for IEEE CIS fraud detection.

    Each item returns:
        tabular_features : (n_features,) float tensor
        time_index       : scalar int (hour of day)
        node_idx         : index into node feature matrix
        label            : binary float {0.0, 1.0}

    Args:s
        tabular_data  : numpy array of tabular features (N, F)
        time_data     : numpy array of time indices (N,)
        labels        : numpy array of fraud labels (N,)
        node_indices  : numpy array of node indices (N,) in graph
    """
    def __init__(self, tabular_data: np.ndarray,
                 time_data: np.ndarray,
                 labels: np.ndarray,
                 node_indices: np.ndarray):

        self.tabular  = torch.FloatTensor(tabular_data)   # (N, F)
        self.time     = torch.LongTensor(time_data)        # (N,)
        self.labels   = torch.FloatTensor(labels)          # (N,)
        self.node_idx = torch.LongTensor(node_indices)     # (N,)

    def __len__(self) -> int:
        """Return number of samples in dataset."""
        return len(self.labels)

    def __getitem__(self, idx: int) -> tuple:
        """
        Return single sample by index.

        Returns:
            tuple: (tabular, time, node_idx, label)
        """
        return (
            self.tabular[idx],
            self.time[idx],
            self.node_idx[idx],
            self.labels[idx],
        )


# ────────────────────────────────────────────────────────────
def prepare_data(tabular_path: str,
                 node_path: str,
                 edge_path: str) -> dict:
    """
    Load and prepare all data splits for training.

    Performs:
      1. Load tabular features and labels
      2. Split into train/val/test (70/15/15)
      3. Standardize features (fit on train only → no leakage)
      4. Load graph node features and edges
      5. Create PyTorch DataLoaders with weighted sampling

    Args:
        tabular_path : path to module3 engineered features CSV
        node_path    : path to module4 node features CSV
        edge_path    : path to module4 edges CSV

    Returns:
        data_dict : dict containing all DataLoaders and metadata
    """
    print("[MODULE 6] ── Preparing Data Splits ──")

    # ── Load tabular features ─────────────────────────────────
    print(f"  ► Loading tabular features from {tabular_path}...")
    df = pd.read_csv(tabular_path)

    # Drop rows where 'isFraud' is NaN, as this is the target variable
    initial_rows = len(df)
    df.dropna(subset=['isFraud'], inplace=True)
    if len(df) < initial_rows:
        print(f"    Dropped {initial_rows - len(df)} rows with NaN in 'isFraud'.")

    # Separate feature columns, time column, and target
    feature_cols = [c for c in df.columns
                    if c not in TABULAR_EXCLUDE and c in df.select_dtypes(include=np.number).columns]

    X = df[feature_cols].values.astype(np.float32)   # (N, F) feature matrix
    y = df['isFraud'].values.astype(np.float32)       # (N,) labels

    # Time index for temporal encoding (use hour_of_day if available)
    if TIME_COL in df.columns:
        t = df[TIME_COL].fillna(0).values.astype(int)
    else:
        t = np.zeros(len(y), dtype=int)               # fallback: all hour=0

    n_features = X.shape[1]
    print(f"    Feature dimensions : {n_features}")
    print(f"    Total samples      : {len(y):,}")
    print(f"    Fraud rate         : {y.mean()*100:.2f}%")

    # ── Train/Val/Test split ──────────────────────────────────
    # First split: separate test set (15%)
    indices = np.arange(len(y))
    idx_trainval, idx_test = train_test_split(
        indices, test_size=TEST_SPLIT, random_state=42, stratify=y
    )

    # Second split: separate validation from train (15% of total = ~17.6% of trainval)
    val_fraction = VAL_SPLIT / (1 - TEST_SPLIT)
    idx_train, idx_val = train_test_split(
        idx_trainval, test_size=val_fraction,
        random_state=42, stratify=y[idx_trainval]
    )

    print(f"    Train: {len(idx_train):,} | Val: {len(idx_val):,} | Test: {len(idx_test):,}")

    # ── Feature standardization ───────────────────────────────
    # IMPORTANT: fit scaler ONLY on training data to prevent data leakage
    scaler = StandardScaler()
    X_train = scaler.fit_transform(X[idx_train])   # fit + transform on train
    X_val   = scaler.transform(X[idx_val])          # transform only (no fit)
    X_test  = scaler.transform(X[idx_test])          # transform only (no fit)

    # ── Load graph data ───────────────────────────────────────
    print(f"  ► Loading graph data...")
    node_df = pd.read_csv(node_path, index_col=0)   # node features with original row indices
    edge_df = pd.read_csv(edge_path)                 # edge list

    # Extract node feature matrix (use available NODE_FEATURE_COLS)
    avail_node_cols = [c for c in NODE_FEATURE_COLS if c in node_df.columns]
    node_features = node_df[avail_node_cols].fillna(0).values.astype(np.float32)

    # Create edge tensors in PyTorch format
    src_nodes  = torch.LongTensor(edge_df['src'].values)    # source node indices
    dst_nodes  = torch.LongTensor(edge_df['dst'].values)    # destination node indices
    edge_index = torch.stack([src_nodes, dst_nodes], dim=0) # (2, E) standard PyG format

    # Convert node features to tensor
    node_feat_tensor = torch.FloatTensor(node_features)     # (N, node_feat_dim)

    print(f"    Graph nodes : {node_feat_tensor.shape[0]:,}")
    print(f"    Graph edges : {edge_index.shape[1]:,}")

    # ── Weighted Sampler for class imbalance ──────────────────
    # Oversamples minority (fraud) class to balance mini-batches
    train_labels = y[idx_train]
    n_fraud  = train_labels.sum()                           # number of fraud cases
    n_legit  = len(train_labels) - n_fraud                  # number of legit cases
    w_fraud  = 1.0 / n_fraud  if n_fraud > 0 else 1.0     # weight for fraud class
    w_legit  = 1.0 / n_legit  if n_legit > 0 else 1.0     # weight for legit class
    sample_weights = torch.FloatTensor([
        w_fraud if label == 1 else w_legit for label in train_labels
    ])
    weighted_sampler = WeightedRandomSampler(
        weights=sample_weights,
        num_samples=len(sample_weights),
        replacement=True         # sampling with replacement for oversampling
    )

    # ── Create PyTorch Datasets ───────────────────────────────
    train_dataset = FraudDataset(X_train, t[idx_train], y[idx_train], idx_train)
    val_dataset   = FraudDataset(X_val,   t[idx_val],   y[idx_val],   idx_val)
    test_dataset  = FraudDataset(X_test,  t[idx_test],  y[idx_test],  idx_test)

    # ── Create DataLoaders ────────────────────────────────────
    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        sampler=weighted_sampler,   # use weighted sampler instead of shuffle
        num_workers=0,              # 0 for Windows compatibility
        pin_memory=True,
    )
    val_loader = DataLoader(
        val_dataset, batch_size=BATCH_SIZE * 2,
        shuffle=False, num_workers=0
    )
    test_loader = DataLoader(
        test_dataset, batch_size=BATCH_SIZE * 2,
        shuffle=False, num_workers=0
    )

    print(f"  ✓ Data preparation complete")

    return {
        'train_loader'    : train_loader,
        'val_loader'      : val_loader,
        'test_loader'     : test_loader,
        'node_features'   : node_feat_tensor,
        'edge_index'      : edge_index,
        'n_tabular_feats' : n_features,
        'n_node_feats'    : node_feat_tensor.shape[1],
        'scaler'          : scaler,
        'idx_test'        : idx_test,
        'y_test'          : y[idx_test],
    }


# ────────────────────────────────────────────────────────────
def train_one_epoch(model, loader, optimizer, node_features,
                    edge_index, device) -> float:
    """
    Execute one full training epoch over all mini-batches.

    For each mini-batch:
      1. Forward pass: compute model outputs
      2. Loss computation: focal loss + proposal loss
      3. Backward pass: compute gradients
      4. Optimizer step: update weights
      5. Gradient clipping: prevent exploding gradients

    Args:
        model          : SpaTempFraudNet instance
        loader         : training DataLoader
        optimizer      : Adam optimizer
        node_features  : full graph node feature tensor (on device)
        edge_index     : full graph edge index tensor (on device)
        device         : CPU or CUDA device

    Returns:
        avg_loss : average training loss for this epoch
    """
    model.train()       # enable dropout, batch norm in training mode
    total_loss = 0.0
    n_batches  = 0

    for tabular, time_idx, node_idx, labels in loader:

        # Move all tensors to device (GPU or CPU)
        tabular  = tabular.to(device)    # (B, F) tabular features
        time_idx = time_idx.to(device)   # (B,) time indices
        labels   = labels.to(device)     # (B,) fraud labels

        # ── Zero gradients from previous step ─────────────────
        optimizer.zero_grad()

        # ── Forward pass ──────────────────────────────────────
        outputs = model(
            tabular_features=tabular,
            node_features=node_features,  # full graph (all N nodes)
            edge_index=edge_index,
            time_indices=time_idx,
        )

        # ── Loss computation ───────────────────────────────────
        loss = model.compute_loss(outputs, labels)

        # ── Backward pass (compute gradients) ─────────────────
        loss.backward()

        # ── Gradient clipping (max norm = 1.0) ────────────────
        # Prevents gradient explosion in deep networks
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

        # ── Optimizer step (update weights) ───────────────────
        optimizer.step()

        # Accumulate loss
        total_loss += loss.item()
        n_batches  += 1

    return total_loss / max(n_batches, 1)


# ────────────────────────────────────────────────────────────
def validate_epoch(model, loader, node_features,
                   edge_index, device) -> tuple:
    """
    Run validation: compute loss and collect predictions.

    Args:
        model, loader, node_features, edge_index, device : same as train_one_epoch

    Returns:
        val_loss  : average validation loss
        all_probs : numpy array of predicted probabilities
        all_labels: numpy array of true labels
    """
    model.eval()        # disable dropout, use running stats for batch norm
    total_loss = 0.0
    all_probs  = []
    all_labels = []

    with torch.no_grad():   # no gradient computation needed during validation
        for tabular, time_idx, node_idx, labels in loader:

            tabular  = tabular.to(device)
            time_idx = time_idx.to(device)
            labels   = labels.to(device)

            outputs  = model(tabular, node_features, edge_index, time_idx)
            loss     = model.compute_loss(outputs, labels)

            total_loss += loss.item()

            # Collect predictions and true labels for metric computation
            all_probs.append(outputs['probs'].cpu().numpy())
            all_labels.append(labels.cpu().numpy())

    avg_loss   = total_loss / max(len(loader), 1)
    all_probs  = np.concatenate(all_probs)    # (N_val,) predicted probs
    all_labels = np.concatenate(all_labels)   # (N_val,) true labels

    return avg_loss, all_probs, all_labels


# ────────────────────────────────────────────────────────────
def run_module6(data_dict: dict = None) -> dict:
    """
    Orchestrates Module 6: full training pipeline.

    Returns:
        training_results : dict with trained model and training log
    """
    print(f"\n[MODULE 6] ── Training Pipeline ──")
    print(f"  Device: {DEVICE}")

    # ── Load data if not passed directly ──────────────────────
    if data_dict is None:
        data_dict = prepare_data(TABULAR_PATH, NODE_PATH, EDGE_PATH)

    # ── Move graph data to device ─────────────────────────────
    node_features = data_dict['node_features'].to(DEVICE)  # full graph nodes
    edge_index    = data_dict['edge_index'].to(DEVICE)     # full graph edges

    # ── Instantiate SpaTempFraudNet ───────────────────────────
    model = SpaTempFraudNet(
        n_tabular_features=data_dict['n_tabular_feats'],
        node_feat_dim=data_dict['n_node_feats'],
        d_model=128,
        gnn_hidden=64,
        gnn_out_dim=128,
        n_atf_heads=8,
        n_atf_layers=3,
        dropout=0.1,
        max_span=50,
    ).to(DEVICE)

    print(f"  Model parameters: {count_parameters(model):,}")

    # ── Optimizer: AdamW ──────────────────────────────────────
    # AdamW = Adam + decoupled weight decay (better for transformers)
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY,
        betas=(0.9, 0.999),
    )

    # ── Learning Rate Scheduler: CosineAnnealingLR ────────────
    # Gradually reduces LR following a cosine curve
    # Prevents oscillation near optimal solution
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=EPOCHS,
        eta_min=1e-6,
    )

    # ── Training Loop ─────────────────────────────────────────
    best_val_loss = float('inf')
    patience_counter = 0
    training_log = []

    print(f"\n  ── Starting Training ({EPOCHS} max epochs) ──")
    print(f"  {'Epoch':>5} {'Train Loss':>12} {'Val Loss':>10} {'LR':>10} {'Status':>10}")
    print("  " + "─" * 55)

    for epoch in range(1, EPOCHS + 1):

        # ── Train one epoch ────────────────────────────────────
        train_loss = train_one_epoch(
            model, data_dict['train_loader'],
            optimizer, node_features, edge_index, DEVICE
        )

        # ── Validate ───────────────────────────────────────────
        val_loss, val_probs, val_labels = validate_epoch(
            model, data_dict['val_loader'],
            node_features, edge_index, DEVICE
        )

        # ── Compute quick validation AUC ───────────────────────
        from sklearn.metrics import roc_auc_score
        try:
            val_auc = roc_auc_score(val_labels, val_probs)
        except Exception:
            val_auc = 0.0

        # ── Step the scheduler ────────────────────────────────
        current_lr = optimizer.param_groups[0]['lr']
        scheduler.step()

        # ── Early stopping check ──────────────────────────────
        status = ''
        if val_loss < best_val_loss:
            best_val_loss    = val_loss
            patience_counter = 0
            status           = '✓ SAVED'
            # Save best model checkpoint
            torch.save({
                'epoch'     : epoch,
                'model_state': model.state_dict(),
                'optimizer_state': optimizer.state_dict(),
                'val_loss'  : val_loss,
                'val_auc'   : val_auc,
            }, MODEL_SAVE)
        else:
            patience_counter += 1
            status = f'patience {patience_counter}/{PATIENCE}'

        # ── Log this epoch's metrics ──────────────────────────
        log_entry = {
            'epoch'      : epoch,
            'train_loss' : round(train_loss, 6),
            'val_loss'   : round(val_loss, 6),
            'val_auc'    : round(val_auc, 4),
            'lr'         : round(current_lr, 8),
        }
        training_log.append(log_entry)

        # Print progress
        print(f"  {epoch:>5} {train_loss:>12.5f} {val_loss:>10.5f} "
              f"{current_lr:>10.2e}  {status}")

        # ── Early stopping ─────────────────────────────────────
        if patience_counter >= PATIENCE:
            print(f"\n  ► Early stopping at epoch {epoch}")
            break

    # ── Save training log to CSV ──────────────────────────────
    log_df = pd.DataFrame(training_log)
    log_df.to_csv(OUTPUT_LOG, index=False)

    print(f"\n  ✓ Training log saved  → {OUTPUT_LOG}")
    print(f"  ✓ Best model saved    → {MODEL_SAVE}")
    print(f"  ✓ Best validation loss: {best_val_loss:.5f}")
    print("[MODULE 6] Complete ✓\n")

    return {
        'model'       : model,
        'training_log': log_df,
        'data_dict'   : data_dict,
        'node_features': node_features,
        'edge_index'  : edge_index,
    }


# ── Execute ──────────────────────────────────────────────────
if __name__ == "__main__":
    training_results = run_module6()
```

⚙️ Module 7: Complete Fraud Evaluation

```python
# ============================================================
# MODULE 7: COMPLETE FRAUD DETECTION EVALUATION
# ============================================================
# PURPOSE  : Evaluate the trained SpaTempFraudNet on the held-out
#            test set using ALL standard fraud detection metrics.
#
# METRICS COMPUTED:
#   ┌─────────────────────────────────────────────────────┐
#   │ ROC-AUC  : Area under ROC curve (threshold-agnostic)│
#   │ PR-AUC   : Area under Precision-Recall curve        │
#   │           (best metric for imbalanced data)         │
#   │ Precision: TP / (TP + FP) — fraud alert quality     │
#   │ Recall   : TP / (TP + FN) — fraud capture rate      │
#   │ F1 Score : Harmonic mean of Precision & Recall       │
#   │ MCC      : Matthews Correlation Coefficient         │
#   │           (best single metric for imbalanced data)  │
#   └─────────────────────────────────────────────────────┘
#
# INPUT    : module6_best_model.pt  (trained model checkpoint)
# OUTPUT   : module7_evaluation_results.xlsx (all metrics)
#            module7_predictions.csv (per-sample probabilities)
#            module7_confusion_matrix.csv
# ============================================================

import torch
import numpy as np
import pandas as pd
from sklearn.metrics import (
    roc_auc_score,             # ROC-AUC: overall discrimination ability
    average_precision_score,   # PR-AUC: performance on imbalanced data
    precision_score,           # TP/(TP+FP): quality of fraud alerts
    recall_score,              # TP/(TP+FN): fraction of fraud caught
    f1_score,                  # harmonic mean of precision & recall
    matthews_corrcoef,         # MCC: correlation between predicted & actual
    confusion_matrix,          # raw TP/TN/FP/FN counts
    classification_report,     # full per-class metric breakdown
    roc_curve,                 # ROC curve points (FPR, TPR, thresholds)
    precision_recall_curve,    # PR curve points (P, R, thresholds)
)
import warnings
warnings.filterwarnings('ignore')

CHECKPOINT_PATH = 'module6_best_model.pt'
OUTPUT_EXCEL    = 'module7_evaluation_results.xlsx'
OUTPUT_PREDS    = 'module7_predictions.csv'
OUTPUT_CM       = 'module7_confusion_matrix.csv'

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Optimal classification threshold (default 0.5; may be tuned)
DEFAULT_THRESHOLD = 0.5


# ────────────────────────────────────────────────────────────
def find_optimal_threshold(y_true: np.ndarray,
                            y_probs: np.ndarray) -> float:
    """
    Find the classification threshold that maximizes F1 score.

    Default threshold of 0.5 is rarely optimal for imbalanced fraud data.
    Scanning all PR curve thresholds finds the F1-maximizing point.

    Math:
        F1(t) = 2 * P(t) * R(t) / (P(t) + R(t))
        t* = argmax_t F1(t)

    Args:
        y_true  : (N,) true binary labels
        y_probs : (N,) predicted fraud probabilities

    Returns:
        optimal_threshold : float threshold that maximizes F1
    """
    # Compute precision and recall at all thresholds
    precisions, recalls, thresholds = precision_recall_curve(y_true, y_probs)

    # Compute F1 at each threshold (avoid division by zero)
    f1_scores = np.where(
        (precisions + recalls) > 0,
        2 * precisions * recalls / (precisions + recalls),
        0
    )

    # Find threshold with maximum F1
    best_idx       = np.argmax(f1_scores)
    best_threshold = thresholds[best_idx] if best_idx < len(thresholds) else 0.5
    best_f1        = f1_scores[best_idx]

    print(f"  ► Optimal threshold: {best_threshold:.4f} (F1={best_f1:.4f})")
    return float(best_threshold)


# ────────────────────────────────────────────────────────────
def collect_test_predictions(model, test_loader,
                              node_features, edge_index,
                              device) -> tuple:
    """
    Run model inference on the full test set.

    Args:
        model          : trained SpaTempFraudNet
        test_loader    : DataLoader for test set
        node_features  : graph node features tensor
        edge_index     : graph edge index tensor
        device         : CPU or CUDA device

    Returns:
        y_true  : (N_test,) true fraud labels
        y_probs : (N_test,) predicted fraud probabilities
    """
    print("  ► Running inference on test set...")

    model.eval()   # inference mode: no dropout
    all_probs  = []
    all_labels = []

    with torch.no_grad():
        for tabular, time_idx, node_idx, labels in test_loader:
            tabular  = tabular.to(device)
            time_idx = time_idx.to(device)

            # Forward pass (no gradient tracking)
            outputs = model(tabular, node_features, edge_index, time_idx)

            # Collect predicted probabilities and true labels
            all_probs.append(outputs['probs'].cpu().numpy())
            all_labels.append(labels.numpy())

    y_true  = np.concatenate(all_labels)   # (N_test,) true labels
    y_probs = np.concatenate(all_probs)    # (N_test,) predicted probs

    print(f"  ► Test set size: {len(y_true):,} | Fraud: {int(y_true.sum()):,}")
    return y_true, y_probs


# ────────────────────────────────────────────────────────────
def compute_all_metrics(y_true: np.ndarray,
                         y_probs: np.ndarray,
                         threshold: float) -> dict:
    """
    Compute the complete set of fraud detection evaluation metrics.

    Args:
        y_true     : (N,) true binary labels {0, 1}
        y_probs    : (N,) predicted fraud probabilities in [0, 1]
        threshold  : classification cutoff (default 0.5, or optimized)

    Returns:
        metrics : dict with all metric names and values

    Metric Interpretations for Fraud Detection:
        ROC-AUC  : 0.5=random, 1.0=perfect. Target: >0.90 for good models.
        PR-AUC   : More informative than ROC for imbalanced data.
                   Baseline = fraud_rate (e.g., 0.035 for 3.5% fraud).
                   Target: >0.80 for excellent fraud models.
        Precision: Of all fraud alerts raised, what fraction are real fraud?
                   Low precision → many false alarms → expensive investigations.
        Recall   : Of all actual fraud cases, what fraction did we catch?
                   Low recall → fraudsters escape detection.
        F1 Score : Balanced metric when precision and recall are equally important.
        MCC      : Best single number for imbalanced binary classification.
                   Ranges from -1 to +1; +1 = perfect, 0 = random.
    """
    # Convert probabilities to binary predictions using threshold
    y_pred = (y_probs >= threshold).astype(int)   # {0, 1}

    # ── ROC-AUC ──────────────────────────────────────────────
    # Area under Receiver Operating Characteristic curve
    # Measures model's ability to discriminate fraud from non-fraud
    # Uses probabilities (not binary predictions) → threshold-agnostic
    roc_auc = roc_auc_score(y_true, y_probs)

    # ── PR-AUC (Average Precision) ────────────────────────────
    # Area under Precision-Recall curve
    # More informative than ROC-AUC for imbalanced datasets
    # Baseline = prevalence rate (3.5% fraud → random model gets ~0.035)
    pr_auc = average_precision_score(y_true, y_probs)

    # ── Precision ─────────────────────────────────────────────
    # Precision = TP / (TP + FP)
    # "Of transactions flagged as fraud, how many are actually fraud?"
    # zero_division=0 returns 0 when denominator is 0
    precision = precision_score(y_true, y_pred, zero_division=0)

    # ── Recall (Sensitivity) ──────────────────────────────────
    # Recall = TP / (TP + FN)
    # "Of all actual fraud cases, what fraction did we detect?"
    recall = recall_score(y_true, y_pred, zero_division=0)

    # ── F1 Score ──────────────────────────────────────────────
    # F1 = 2 * (Precision * Recall) / (Precision + Recall)
    # Harmonic mean balances precision and recall
    f1 = f1_score(y_true, y_pred, zero_division=0)

    # ── MCC (Matthews Correlation Coefficient) ────────────────
    # MCC = (TP*TN - FP*FN) / sqrt((TP+FP)(TP+FN)(TN+FP)(TN+FN))
    # Best single metric for imbalanced binary classification
    # Returns value in [-1, 1]; 0 = random; 1 = perfect
    mcc = matthews_corrcoef(y_true, y_pred)

    # ── Confusion Matrix ──────────────────────────────────────
    # [[TN, FP],
    #  [FN, TP]]
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()

    # ── Derived metrics ───────────────────────────────────────
    # Specificity = TN / (TN + FP) — true negative rate
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0

    # False Positive Rate = FP / (FP + TN)
    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0

    # False Discovery Rate = FP / (FP + TP)
    fdr = fp / (fp + tp) if (fp + tp) > 0 else 0.0

    metrics = {
        # Primary metrics
        'ROC_AUC'          : round(roc_auc, 4),
        'PR_AUC'           : round(pr_auc, 4),
        'Precision'        : round(precision, 4),
        'Recall'           : round(recall, 4),
        'F1_Score'         : round(f1, 4),
        'MCC'              : round(mcc, 4),
        # Raw counts
        'True_Positives'   : int(tp),
        'True_Negatives'   : int(tn),
        'False_Positives'  : int(fp),
        'False_Negatives'  : int(fn),
        # Derived metrics
        'Specificity'      : round(specificity, 4),
        'False_Pos_Rate'   : round(fpr, 4),
        'False_Disc_Rate'  : round(fdr, 4),
        # Configuration
        'Threshold'        : round(threshold, 4),
        'Test_Size'        : len(y_true),
        'Fraud_Count'      : int(y_true.sum()),
        'Legit_Count'      : int((y_true == 0).sum()),
    }

    return metrics, cm


# ────────────────────────────────────────────────────────────
def print_metrics_table(metrics: dict):
    """Pretty-print all metrics to console."""
    print("\n  ═══ EVALUATION RESULTS ════════════════════════")
    print(f"  {'Metric':<25} {'Value':>10}")
    print("  " + "─" * 40)
    print(f"  {'ROC-AUC':<25} {metrics['ROC_AUC']:>10.4f}")
    print(f"  {'PR-AUC':<25} {metrics['PR_AUC']:>10.4f}")
    print(f"  {'Precision':<25} {metrics['Precision']:>10.4f}")
    print(f"  {'Recall (Sensitivity)':<25} {metrics['Recall']:>10.4f}")
    print(f"  {'F1 Score':<25} {metrics['F1_Score']:>10.4f}")
    print(f"  {'MCC':<25} {metrics['MCC']:>10.4f}")
    print(f"  {'Specificity':<25} {metrics['Specificity']:>10.4f}")
    print("  " + "─" * 40)
    print(f"  {'True  Positives (TP)':<25} {metrics['True_Positives']:>10,}")
    print(f"  {'True  Negatives (TN)':<25} {metrics['True_Negatives']:>10,}")
    print(f"  {'False Positives (FP)':<25} {metrics['False_Positives']:>10,}")
    print(f"  {'False Negatives (FN)':<25} {metrics['False_Negatives']:>10,}")
    print("  ══════════════════════════════════════════════\n")


# ────────────────────────────────────────────────────────────
def save_evaluation_outputs(metrics: dict, cm: np.ndarray,
                             y_true: np.ndarray, y_probs: np.ndarray,
                             idx_test: np.ndarray):
    """
    Save all evaluation outputs to CSV and Excel.

    Saves:
      1. Excel with metrics, confusion matrix, ROC/PR curve data
      2. CSV with per-transaction predictions and probabilities
      3. CSV with raw confusion matrix
    """
    print("  ► Saving evaluation outputs...")

    # ── 1. Per-sample predictions CSV ─────────────────────────
    preds_df = pd.DataFrame({
        'transaction_index' : idx_test,               # original row index
        'true_label'        : y_true.astype(int),     # 0=legit, 1=fraud
        'fraud_probability' : y_probs.round(6),       # model confidence score
        'predicted_label'   : (y_probs >= metrics['Threshold']).astype(int),
        'correct'           : (
            (y_probs >= metrics['Threshold']).astype(int) == y_true.astype(int)
        ).astype(int),
    })
    preds_df.to_csv(OUTPUT_PREDS, index=False)

    # ── 2. Confusion matrix CSV ────────────────────────────────
    cm_df = pd.DataFrame(
        cm,
        index=['Actual Legit (0)', 'Actual Fraud (1)'],
        columns=['Predicted Legit (0)', 'Predicted Fraud (1)']
    )
    cm_df.to_csv(OUTPUT_CM)

    # ── 3. Compute ROC curve data ─────────────────────────────
    fpr_vals, tpr_vals, roc_thresh = roc_curve(y_true, y_probs)
    roc_curve_df = pd.DataFrame({
        'False_Positive_Rate': fpr_vals,
        'True_Positive_Rate' : tpr_vals,
        'Threshold'          : np.append(roc_thresh, np.nan),
    })

    # ── 4. Compute PR curve data ──────────────────────────────
    prec_vals, rec_vals, pr_thresh = precision_recall_curve(y_true, y_probs)
    pr_curve_df = pd.DataFrame({
        'Precision' : prec_vals,
        'Recall'    : rec_vals,
        'Threshold' : np.append(pr_thresh, np.nan),
    })

    # ── 5. Save all to Excel (multi-sheet) ───────────────────
    metrics_df = pd.DataFrame([metrics]).T.reset_index()
    metrics_df.columns = ['Metric', 'Value']

    with pd.ExcelWriter(OUTPUT_EXCEL, engine='openpyxl') as writer:
        metrics_df.to_excel(writer,   sheet_name='All_Metrics',       index=False)
        cm_df.to_excel(writer,        sheet_name='Confusion_Matrix')
        roc_curve_df.to_excel(writer, sheet_name='ROC_Curve_Data',    index=False)
        pr_curve_df.to_excel(writer,  sheet_name='PR_Curve_Data',     index=False)
        preds_df.head(10000).to_excel(writer, sheet_name='Sample_Predictions', index=False)

    print(f"  ✓ Evaluation Excel   → {OUTPUT_EXCEL}")
    print(f"  ✓ Predictions CSV    → {OUTPUT_PREDS}")
    print(f"  ✓ Confusion Matrix   → {OUTPUT_CM}")


# ────────────────────────────────────────────────────────────
def run_module7(training_results: dict = None):
    """
    Orchestrates Module 7: complete evaluation pipeline.

    Args:
        training_results : dict from run_module6() with model and data

    Returns:
        metrics : dict with all evaluation metrics
    """
    print(f"\n[MODULE 7] ── Fraud Detection Evaluation ──")

    if training_results is None:
        print("  [!] training_results not provided.")
        print("  Please pass the output of run_module6() to run_module7()")
        return None

    model         = training_results['model']
    data_dict     = training_results['data_dict']
    node_features = training_results['node_features']
    edge_index    = training_results['edge_index']

    # ── Load best checkpoint ──────────────────────────────────
    print(f"  ► Loading best checkpoint from {CHECKPOINT_PATH}...")
    checkpoint = torch.load(CHECKPOINT_PATH, map_location=DEVICE)
    model.load_state_dict(checkpoint['model_state'])
    print(f"    Best epoch: {checkpoint['epoch']} | Val AUC: {checkpoint['val_auc']:.4f}")

    # ── Collect test predictions ──────────────────────────────
    y_true, y_probs = collect_test_predictions(
        model, data_dict['test_loader'],
        node_features, edge_index, DEVICE
    )

    # ── Find optimal threshold ────────────────────────────────
    optimal_threshold = find_optimal_threshold(y_true, y_probs)

    # ── Compute all metrics ───────────────────────────────────
    print("\n  ► Computing evaluation metrics at optimal threshold...")
    metrics, cm = compute_all_metrics(y_true, y_probs, optimal_threshold)

    # Print to console
    print_metrics_table(metrics)

    # ── Save outputs ──────────────────────────────────────────
    save_evaluation_outputs(
        metrics, cm, y_true, y_probs,
        data_dict['idx_test']
    )

    print("[MODULE 7] Complete ✓\n")
    return metrics


# ── Execute ──────────────────────────────────────────────────
if __name__ == "__main__":
    # In practice: training_results = run_module6()
    #              metrics = run_module7(training_results)
    print("Run this module with training_results from Module 6.")
```

⚙️ Module 8: Explainability (SHAP)

```python
# ============================================================
# MODULE 8: EXPLAINABILITY WITH SHAP
# ============================================================
# PURPOSE  : Generate SHAP (SHapley Additive exPlanations) values
#            to explain individual fraud predictions.
#            This supports the "LLM-Based Trustworthy Explanations"
#            part of your thesis title.
# INPUT    : trained model, test data
# OUTPUT   : module8_shap_values.csv
#            module8_shap_summary.xlsx
# ============================================================

import torch
import numpy as np
import pandas as pd
import shap
import warnings
warnings.filterwarnings('ignore')

OUTPUT_SHAP    = 'module8_shap_values.csv'
OUTPUT_SUMMARY = 'module8_shap_summary.xlsx'
N_EXPLAIN      = 500    # number of test samples to explain (SHAP is slow)
N_BACKGROUND   = 100    # background samples for SHAP KernelExplainer


# ────────────────────────────────────────────────────────────
def create_model_wrapper(model, node_features, edge_index,
                          device, time_col_idx: int = 0):
    """
    Create a function wrapper around the model for SHAP compatibility.

    SHAP KernelExplainer requires a function f(X) → predictions.
    This wrapper handles the model's multi-input architecture
    by fixing graph data and only varying tabular features.

    Args:
        model        : trained SpaTempFraudNet
        node_features: graph node features (fixed for explanation)
        edge_index   : graph edges (fixed)
        device       : CPU or CUDA
        time_col_idx : index of time column in tabular features

    Returns:
        predict_fn : callable f(X_numpy) → probabilities (N,)
    """
    model.eval()

    def predict_fn(X_numpy: np.ndarray) -> np.ndarray:
        """
        Model prediction function for SHAP.

        Args:
            X_numpy : (N, n_features) numpy array of tabular features

        Returns:
            probs : (N,) numpy array of fraud probabilities
        """
        # Convert numpy array to tensor
        X_tensor = torch.FloatTensor(X_numpy).to(device)

        # Extract time index from features (column at time_col_idx)
        # Clip to valid range [0, 23] for hour encoding
        time_idx = X_tensor[:, time_col_idx].long().clamp(0, 23)

        with torch.no_grad():
            outputs = model(
                tabular_features=X_tensor,
                node_features=node_features,
                edge_index=edge_index,
                time_indices=time_idx,
            )
        # Return numpy array of fraud probabilities
        return outputs['probs'].cpu().numpy()

    return predict_fn


# ────────────────────────────────────────────────────────────
def compute_shap_values(predict_fn,
                         X_train_sample: np.ndarray,
                         X_explain: np.ndarray,
                         feature_names: list) -> np.ndarray:
    """
    Compute SHAP values using KernelExplainer.

    KernelExplainer is model-agnostic — works with any model.
    It approximates Shapley values using a weighted linear model
    over feature subsets.

    Math:
        φ_i = Σ_{S ⊆ F\{i}} [|S|!(|F|-|S|-1)!/|F|!] * [f(S∪{i}) - f(S)]

    where φ_i = SHAP value for feature i (its contribution to prediction)

    Args:
        predict_fn     : model prediction function f(X) → probs
        X_train_sample : (n_background, n_features) background dataset
                         Used as baseline/reference for Shapley computation
        X_explain      : (n_explain, n_features) samples to explain
        feature_names  : list of feature column names

    Returns:
        shap_values : (n_explain, n_features) numpy array of SHAP values
    """
    print(f"  ► Computing SHAP values for {len(X_explain)} samples...")
    print(f"    Background samples: {len(X_train_sample)}")
    print(f"    (This may take several minutes...)")

    # Create SHAP KernelExplainer with background dataset
    # Background data represents the "baseline" (expected) prediction
    explainer = shap.KernelExplainer(
        model=predict_fn,          # model prediction function
        data=X_train_sample,       # background/baseline samples
        link='identity',           # link function: raw probabilities (not logit)
    )

    # Compute SHAP values for explanation samples
    # nsamples: number of perturbations per sample (more = more accurate but slower)
    shap_values = explainer.shap_values(
        X_explain,
        nsamples=100,    # 100 permutations per sample (balance speed vs accuracy)
    )

    print(f"  ✓ SHAP values computed. Shape: {shap_values.shape}")
    return shap_values


# ────────────────────────────────────────────────────────────
def save_shap_outputs(shap_values: np.ndarray,
                       X_explain: np.ndarray,
                       y_explain: np.ndarray,
                       feature_names: list):
    """
    Save SHAP values and summary statistics to CSV/Excel.

    SHAP value interpretation:
        φ_i > 0 → feature i INCREASES fraud probability
        φ_i < 0 → feature i DECREASES fraud probability
        |φ_i|   → magnitude of feature i's influence on this prediction

    Args:
        shap_values   : (N, F) SHAP values matrix
        X_explain     : (N, F) original feature values
        y_explain     : (N,) true labels
        feature_names : list of F feature names
    """
    print("  ► Saving SHAP outputs...")

    # ── Per-sample SHAP values CSV ─────────────────────────────
    shap_df = pd.DataFrame(shap_values, columns=[f'shap_{c}' for c in feature_names])
    shap_df.insert(0, 'true_label', y_explain.astype(int))
    shap_df.to_csv(OUTPUT_SHAP, index=False)

    # ── Feature importance summary ─────────────────────────────
    # Mean absolute SHAP value per feature = global feature importance
    mean_abs_shap = np.abs(shap_values).mean(axis=0)   # (F,)

    # Mean SHAP value per feature (positive = generally increases fraud risk)
    mean_shap = shap_values.mean(axis=0)               # (F,)

    summary_df = pd.DataFrame({
        'feature'           : feature_names,
        'mean_abs_shap'     : mean_abs_shap.round(6),  # global importance
        'mean_shap'         : mean_shap.round(6),      # directional impact
        'pct_positive_shap' : (shap_values > 0).mean(axis=0).round(4),  # how often ↑ risk
    }).sort_values('mean_abs_shap', ascending=False)  # rank by importance

    # Print top 15 most important features
    print("\n  ── Top 15 Most Important Features (SHAP) ────────")
    print(f"  {'Rank':<5} {'Feature':<30} {'Importance':>12} {'Direction':>10}")
    print("  " + "─" * 62)
    for rank, (_, row) in enumerate(summary_df.head(15).iterrows(), 1):
        direction = '↑ Risk' if row['mean_shap'] > 0 else '↓ Risk'
        print(f"  {rank:<5} {row['feature']:<30} {row['mean_abs_shap']:>12.4f} {direction:>10}")

    # Save to Excel with two sheets
    with pd.ExcelWriter(OUTPUT_SUMMARY, engine='openpyxl') as writer:
        summary_df.to_excel(writer, sheet_name='Feature_Importance', index=False)

        # Separate sheets for fraud vs legit SHAP values
        fraud_idx = y_explain == 1
        legit_idx = y_explain == 0

        fraud_shap_mean = pd.DataFrame({
            'feature'   : feature_names,
            'mean_shap_fraud': shap_values[fraud_idx].mean(axis=0).round(6),
            'mean_shap_legit': shap_values[legit_idx].mean(axis=0).round(6),
        }).sort_values('mean_shap_fraud', ascending=False)

        fraud_shap_mean.to_excel(writer, sheet_name='Fraud_vs_Legit_SHAP', index=False)

    print(f"\n  ✓ SHAP values CSV   → {OUTPUT_SHAP}")
    print(f"  ✓ SHAP summary Excel → {OUTPUT_SUMMARY}")


# ────────────────────────────────────────────────────────────
def run_module8(training_results: dict = None,
                feature_cols: list = None):
    """
    Orchestrates Module 8: SHAP explainability pipeline.

    Args:
        training_results : dict from run_module6()
        feature_cols     : list of tabular feature column names
    """
    print(f"\n[MODULE 8] ── SHAP Explainability ──")

    if training_results is None:
        print("  [!] Pass training_results from run_module6()")
        return None

    model         = training_results['model']
    data_dict     = training_results['data_dict']
    node_features = training_results['node_features']
    edge_index    = training_results['edge_index']

    # Create model prediction wrapper for SHAP
    predict_fn = create_model_wrapper(
        model, node_features, edge_index, DEVICE, time_col_idx=0
    )

    # ── Sample data for SHAP analysis ─────────────────────────
    print(f"  ► Preparing {N_EXPLAIN} samples for explanation...")

    # Collect samples from test loader
    all_X, all_y = [], []
    for tabular, time_idx, node_idx, labels in data_dict['test_loader']:
        all_X.append(tabular.numpy())
        all_y.append(labels.numpy())
        if sum(len(x) for x in all_X) >= N_EXPLAIN + N_BACKGROUND:
            break

    all_X = np.vstack(all_X)
    all_y = np.concatenate(all_y)

    # Background samples (first N_BACKGROUND, random selection in practice)
    X_background = all_X[:N_BACKGROUND]

    # Explanation samples (next N_EXPLAIN)
    X_explain = all_X[N_BACKGROUND : N_BACKGROUND + N_EXPLAIN]
    y_explain = all_y[N_BACKGROUND : N_BACKGROUND + N_EXPLAIN]

    # ── Compute SHAP values ────────────────────────────────────
    shap_values = compute_shap_values(
        predict_fn, X_background, X_explain,
        feature_names=feature_cols or [f'feat_{i}' for i in range(X_explain.shape[1])]
    )

    # ── Save outputs ──────────────────────────────────────────
    save_shap_outputs(
        shap_values, X_explain, y_explain,
        feature_names=feature_cols or [f'feat_{i}' for i in range(X_explain.shape[1])]
    )

    print("[MODULE 8] Complete ✓\n")
    return shap_values
```
